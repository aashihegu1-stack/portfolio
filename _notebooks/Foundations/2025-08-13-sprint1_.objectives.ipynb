```python
<!DOCTYPE html>
<html lang="en"><head>
  <meta charset="utf-8">
  <meta http-equiv="X-UA-Compatible" content="IE=edge">
  <meta name="viewport" content="width=device-width, initial-scale=1"><!-- Begin Jekyll SEO tag v2.9.0 -->
<title>Onboarding Challenge - Ground 0 | Open Coding</title>
<meta name="generator" content="Jekyll v3.9.5" />
<meta property="og:title" content="Onboarding Challenge - Ground 0" />
<meta property="og:locale" content="en_US" />
<meta name="description" content="Master essential development tools and complete team onboarding through hands-on setup, coding, and collaboration. Begin the process for Project-based Learning (PBL) by launching into a project." />
<meta name="twitter:description" property="og:description" content="Master essential development tools and complete team onboarding through hands-on setup, coding, and collaboration. Begin the process for Project-based Learning (PBL) by launching into a project." />
<meta property="og:site_name" content="Open Coding" />
<meta property="og:type" content="article" />
<meta property="article:published_time" content="2025-08-22T00:00:00+00:00" />
<meta property="article:modified_time" content="2025-08-22T00:00:00+00:00" />
<meta name="twitter:card" content="summary" />
<meta name="twitter:title" content="Onboarding Challenge - Ground 0" />
<script type="application/ld+json">
{"@context":"https://schema.org","@type":"BlogPosting","dateModified":"2025-08-22T00:00:00+00:00","datePublished":"2025-08-22T00:00:00+00:00","description":"Master essential development tools and complete team onboarding through hands-on setup, coding, and collaboration. Begin the process for Project-based Learning (PBL) by launching into a project.","headline":"Onboarding Challenge - Ground 0","mainEntityOfPage":{"@type":"WebPage","@id":"/sprint1/challenge/csp/"},"url":"/sprint1/challenge/csp/"}</script>
<!-- End Jekyll SEO tag -->
<link id="main-stylesheet" rel="stylesheet" href="/assets/css/style.css"><link type="application/atom+xml" rel="alternate" href="/feed.xml" title="Open Coding" />
<!-- Minima start custom head snippets -->
<!-- Favicon for browser tabs and bookmarks -->
<link rel="icon" href="/favicon.ico" type="image/x-icon">

<!-- Support for Apple devices (iPhone/iPad homescreen icon) -->
<link rel="apple-touch-icon" sizes="180x180" href="/apple-touch-icon.png">

<!-- Support for modern browsers and pinned tabs -->
<link rel="icon" type="image/png" sizes="32x32" href="/favicon-32x32.png">
<link rel="icon" type="image/png" sizes="16x16" href="/favicon-16x16.png">

<!-- Google Fonts - commonly used across the site -->
<link rel="preconnect" href="https://fonts.googleapis.com">
<link rel="preconnect" href="https://fonts.gstatic.com" crossorigin>
<link href="https://fonts.googleapis.com/css2?family=Inter:wght@400;500;600;700&family=Roboto:wght@400;500;700&family=Open+Sans:wght@400;600&family=Lato:wght@400;700&family=Montserrat:wght@400;600&family=Noto+Sans:wght@400;700&family=Merriweather:wght@400;700&family=Source+Code+Pro:wght@400;700&family=Orbitron:wght@400;700;900&family=Rajdhani:wght@300;400;600;700&family=Press+Start+2P&display=swap" rel="stylesheet"><link rel="stylesheet" href="https://cdn.datatables.net/1.13.6/css/jquery.dataTables.min.css">
	<script src="https://code.jquery.com/jquery-3.7.1.min.js"></script>
	<script src="https://cdn.datatables.net/1.13.6/js/jquery.dataTables.min.js"></script><link rel="stylesheet" href="/assets/css/style.css">
<script src="/assets/js/user-preferences.js"></script>
<!-- Set up javaURI globally for OCS Analytics Tracker -->
<script>
    if (location.hostname === "localhost" || location.hostname === "127.0.0.1") {
        window.javaURI = "http://localhost:8585";
    } else {
        window.javaURI = "https://spring.opencodingsociety.com";
    }
</script>
<!-- OCS Analytics Tracking Script (loaded on every page) -->
<script src="/assets/js/ocs-analytics-tracker.js" defer></script>
<!-- Keyboard Shortcuts (Alt+Shift+key navigation, loaded on every page) -->
<script>window.baseurl = "";</script>
<script src="/assets/js/keyboard-shortcuts.js" defer></script>

<!-- AI A/B Test Engine SDK (passes site.baseurl for asset resolution) -->
<script src="https://ai-ab-test-engine.vercel.app/sdk.js"
  data-site-baseurl=""
  data-project-id="cmnfour780000f63gu69kg2v8"></script>
<script type="importmap">
{
  "imports": {
    "@assets/": "/assets/",
    "@fortuneFinders/": "/assets/js/fortuneFinders/"
  }
}
</script>

<!-- Minima end custom head snippets -->
 </head>
<body>    <header class="site-header">
        <div class="wrapper"><a class="site-title" rel="author" href="/" data-value="Open Coding">
                <img id="site-logo" src="/images/logo.png" alt="Logo">
            </a><nav class="site-nav">
                <input type="checkbox" id="nav-trigger" class="nav-trigger" />
                <label for="nav-trigger">
                    <span class="menu-icon">
                        <svg viewBox="0 0 18 15" width="18px" height="15px">
                            <path
                                d="M18,1.484c0,0.82-0.665,1.484-1.484,1.484H1.484C0.665,2.969,0,2.304,0,1.484l0,0C0,0.665,0.665,0,1.484,0 h15.032C17.335,0,18,0.665,18,1.484L18,1.484z M18,7.516C18,8.335,17.335,9,16.516,9H1.484C0.665,9,0,8.335,0,7.516l0,0 c0-0.82,0.665-1.484,1.484-1.484h15.032C17.335,6.031,18,6.696,18,7.516L18,7.516z M18,13.516C18,14.335,17.335,15,16.516,15H1.484 C0.665,15,0,14.335,0,13.516l0,0c0-0.82,0.665-1.483,1.484-1.483h15.032C17.335,12.031,18,12.695,18,13.516L18,13.516z" />
                        </svg>
                    </span>
                </label>

                <div class="trigger"><a class="page-link" href="/navigation/blogs/">Blogs</a><a class="page-link" href="/search/">Search</a><a class="page-link" href="/capstone/">Capstone</a>
                    <a id="loginArea" class="page-link" href="/login">Login</a>
                </div>
            </nav></div>
    </header>

    <script type="module" src="/assets/js/api/login.js"></script>    


        

        

        <main class="page-content" aria-label="Content">
            <div class="wrapper">
                                <div class="opencs_root">
                                    















<link rel="stylesheet" href="https://cdnjs.cloudflare.com/ajax/libs/font-awesome/6.5.0/css/all.min.css">

<div class="lesson-player">
  <!-- LEFT SIDEBAR -->
  <!-- LESSON PLAYER SIDEBAR NAVIGATION -->


<div class="lesson-sidebar">
  <div class="sidebar-header">
    
      <h2>Computer Science Principles</h2>
    

    <div class="course-progress-section">
      <div class="progress-label">
        <span>Course Progress</span>
        <span id="progress-count">0/0</span>
      </div>
      <div class="progress-bar-sidebar">
        <div class="progress-bar-sidebar-fill" id="sidebar-progress" style="width: 0%"></div>
      </div>
    </div>
  </div>

  <div class="sprint-nav"><div class="sprint-section">
        <button class="sprint-toggle" onclick="toggleSprint('1')" type="button">
          <i class="fas fa-chevron-right sprint-chevron" id="chevron-1"></i>
          <div class="sprint-info">
            <div class="sprint-title">Sprint1: </div>
            <div class="sprint-weeks">Weeks 0-4</div>
          </div>
        </button>

        <div class="lesson-list" id="sprint-1-lessons" style="display: none;"><a href="/csp/home-page-game-feedback"
                 class="lesson-item "
                 data-lesson-id="/csp/home-page-game-feedback">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Home Page Game Feedback</span>
              </a><a href="/python/intro"
                 class="lesson-item "
                 data-lesson-id="/python/intro">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Intro to Python</span>
              </a><a href="/javascript/intro/csp/"
                 class="lesson-item "
                 data-lesson-id="/javascript/intro/csp/">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Intro to JavaScript</span>
              </a><a href="/csp/cpt-concepts"
                 class="lesson-item "
                 data-lesson-id="/csp/cpt-concepts">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Create Performance Task Concepts</span>
              </a><a href="/tools/csp/"
                 class="lesson-item "
                 data-lesson-id="/tools/csp/">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Tools and Equipment Setup</span>
              </a><a href="/csp/big-idea/fundamentals"
                 class="lesson-item "
                 data-lesson-id="/csp/big-idea/fundamentals">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">BI 3 Review</span>
              </a><a href="/code/csp/"
                 class="lesson-item "
                 data-lesson-id="/code/csp/">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Code Runners - Raw MD</span>
              </a><a href="/code/ui/csp/"
                 class="lesson-item "
                 data-lesson-id="/code/ui/csp/">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">UI Runner</span>
              </a><a href="/rpg/game/csp/"
                 class="lesson-item "
                 data-lesson-id="/rpg/game/csp/">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Game Runner Examples</span>
              </a><a href="/python/variables"
                 class="lesson-item "
                 data-lesson-id="/python/variables">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">3.1 Variables and Assignments</span>
              </a><a href="/code/javascript/csp/"
                 class="lesson-item "
                 data-lesson-id="/code/javascript/csp/">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Code Runner - JavaScript</span>
              </a><a href="/csp/big-idea-3/developing-procedures/p4/introduction"
                 class="lesson-item "
                 data-lesson-id="/csp/big-idea-3/developing-procedures/p4/introduction">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">3.13 Developing Procedures</span>
              </a><a href="/csp/developing-algorithms/crashers"
                 class="lesson-item "
                 data-lesson-id="/csp/developing-algorithms/crashers">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">3.9 Developing Algorithms</span>
              </a><a href="/csp/big-idea-3/calling-procedures/p4/lesson"
                 class="lesson-item "
                 data-lesson-id="/csp/big-idea-3/calling-procedures/p4/lesson">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">3.12 Calling Procedures</span>
              </a><a href="/python/boolean/py"
                 class="lesson-item "
                 data-lesson-id="/python/boolean/py">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">3.5 Boolean Expressions (PY)</span>
              </a><a href="/javascript/boolean/js"
                 class="lesson-item "
                 data-lesson-id="/javascript/boolean/js">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">3.5 Boolean Expressions (JS)</span>
              </a><a href="/peppa-maze/algorithmic-efficiency"
                 class="lesson-item "
                 data-lesson-id="/peppa-maze/algorithmic-efficiency">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">3.17 Algorithmic Efficiency</span>
              </a><a href="/csp/big-idea-3/lists/p4/lesson"
                 class="lesson-item "
                 data-lesson-id="/csp/big-idea-3/lists/p4/lesson">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">3.10 Lists</span>
              </a><a href="/3.15/level1-pseudocode"
                 class="lesson-item "
                 data-lesson-id="/3.15/level1-pseudocode">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">3.15 Random Numbers (pseudocode)</span>
              </a><a href="/csp/strings/zombies"
                 class="lesson-item "
                 data-lesson-id="/csp/strings/zombies">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">3.4 Strings</span>
              </a><a href="/csp/crowdsourcing"
                 class="lesson-item "
                 data-lesson-id="/csp/crowdsourcing">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Crowdsourcing Lesson</span>
              </a><a href="/csp/big-idea-3/scratchers/iterations/p4/lesson"
                 class="lesson-item "
                 data-lesson-id="/csp/big-idea-3/scratchers/iterations/p4/lesson">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">3.8 Iterations</span>
              </a><a href="/csp/codemaxxers/nested-conditionals-lesson"
                 class="lesson-item "
                 data-lesson-id="/csp/codemaxxers/nested-conditionals-lesson">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">3.7 Nested Conditionals</span>
              </a><a href="/csp/big-idea-3/p3/math-expressions/lesson"
                 class="lesson-item "
                 data-lesson-id="/csp/big-idea-3/p3/math-expressions/lesson">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">3.3 Math Expressions</span>
              </a><a href="/csp/big-idea-3/data-abstractions/p4/introduction"
                 class="lesson-item "
                 data-lesson-id="/csp/big-idea-3/data-abstractions/p4/introduction">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">3.2 Data Abstractions</span>
              </a><a href="/csp/big-idea-3/RandomPY/p3/Lesson"
                 class="lesson-item "
                 data-lesson-id="/csp/big-idea-3/RandomPY/p3/Lesson">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">3.15 Random Numbers (py)</span>
              </a><a href="/github/pages/about/csp/"
                 class="lesson-item "
                 data-lesson-id="/github/pages/about/csp/">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">GitHub Pages (about.md)</span>
              </a><a href="/sprint1/challenge/csp/"
                 class="lesson-item active"
                 data-lesson-id="/sprint1/challenge/csp/">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Onboarding Challenge - Ground 0</span>
              </a><a href="/agile/csp/"
                 class="lesson-item "
                 data-lesson-id="/agile/csp/">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Agile Methodologies Practitioner</span>
              </a><a href="/github/pages/csp/"
                 class="lesson-item "
                 data-lesson-id="/github/pages/csp/">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">GitHub Pages (index.md)</span>
              </a><a href="/csp/sprint2/objectives"
                 class="lesson-item "
                 data-lesson-id="/csp/sprint2/objectives">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Objectives in LxD</span>
              </a></div>
      </div><div class="sprint-section">
        <button class="sprint-toggle" onclick="toggleSprint('2')" type="button">
          <i class="fas fa-chevron-right sprint-chevron" id="chevron-2"></i>
          <div class="sprint-info">
            <div class="sprint-title">Sprint2: Full Stack Foundations</div>
            <div class="sprint-weeks">Weeks 5-8</div>
          </div>
        </button>

        <div class="lesson-list" id="sprint-2-lessons" style="display: none;"><a href="/python/flask/crash-course"
                 class="lesson-item "
                 data-lesson-id="/python/flask/crash-course">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Crash Course into Python Flask</span>
              </a><a href="/python/flask/crash-course"
                 class="lesson-item "
                 data-lesson-id="/python/flask/crash-course">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Anatomy of Python Flask</span>
              </a><a href="/csp/sprint3/objectives"
                 class="lesson-item "
                 data-lesson-id="/csp/sprint3/objectives">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Frontend Development Objectives</span>
              </a><a href="/english/plagiarism/storyboard/"
                 class="lesson-item "
                 data-lesson-id="/english/plagiarism/storyboard/">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Plagiarism Prevention Prototype (Sample)</span>
              </a><a href="/sprints/csp/"
                 class="lesson-item "
                 data-lesson-id="/sprints/csp/">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Framework for Sprints</span>
              </a></div>
      </div><div class="sprint-section">
        <button class="sprint-toggle" onclick="toggleSprint('3')" type="button">
          <i class="fas fa-chevron-right sprint-chevron" id="chevron-3"></i>
          <div class="sprint-info">
            <div class="sprint-title">Sprint3: Building a Full Stack Web Application</div>
            <div class="sprint-weeks">Weeks 9-12</div>
          </div>
        </button>

        <div class="lesson-list" id="sprint-3-lessons" style="display: none;"><a href="/plagiarism/tdd-plan"
                 class="lesson-item "
                 data-lesson-id="/plagiarism/tdd-plan">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Plagiarism TDD plan and ceremony (Sample)</span>
              </a><a href="/plagiarism/useability"
                 class="lesson-item "
                 data-lesson-id="/plagiarism/useability">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Plagiarism Usability Test Plan (Sample)</span>
              </a><a href="/flask-flocker-api"
                 class="lesson-item "
                 data-lesson-id="/flask-flocker-api">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Planning Workspaces and APIs (Flocker)</span>
              </a></div>
      </div><div class="sprint-section">
        <button class="sprint-toggle" onclick="toggleSprint('4')" type="button">
          <i class="fas fa-chevron-right sprint-chevron" id="chevron-4"></i>
          <div class="sprint-info">
            <div class="sprint-title">Sprint4: CPT Passion Project: Full Stack Development</div>
            <div class="sprint-weeks">Weeks 13-16</div>
          </div>
        </button>

        <div class="lesson-list" id="sprint-4-lessons" style="display: none;"><a href="/cpt/cpt-guide"
                 class="lesson-item "
                 data-lesson-id="/cpt/cpt-guide">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">CPT Preparation Guide</span>
              </a><a href="/game/essentials/variables/csp/"
                 class="lesson-item "
                 data-lesson-id="/game/essentials/variables/csp/">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Variables and HTML DOM</span>
              </a><a href="/csp/sprint4/objectives"
                 class="lesson-item "
                 data-lesson-id="/csp/sprint4/objectives">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">UX Development Objectives</span>
              </a><a href="/python/flask/csp/"
                 class="lesson-item "
                 data-lesson-id="/python/flask/csp/">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Python/Flask Backend Infograph</span>
              </a><a href="/flask-intro"
                 class="lesson-item "
                 data-lesson-id="/flask-intro">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Intro to Flask</span>
              </a><a href="/flask-debug-walk"
                 class="lesson-item "
                 data-lesson-id="/flask-debug-walk">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Flask Debug Walk-Through</span>
              </a><a href="/flask-code-login"
                 class="lesson-item "
                 data-lesson-id="/flask-code-login">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Frontend Authentication with a Backend Flask Application</span>
              </a><a href="/sprint4/retrospective"
                 class="lesson-item "
                 data-lesson-id="/sprint4/retrospective">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Team Review and Video</span>
              </a></div>
      </div><div class="sprint-section">
        <button class="sprint-toggle" onclick="toggleSprint('5')" type="button">
          <i class="fas fa-chevron-right sprint-chevron" id="chevron-5"></i>
          <div class="sprint-info">
            <div class="sprint-title">Sprint5: CPT Completion and Industry Transition</div>
            <div class="sprint-weeks">Weeks 17-20</div>
          </div>
        </button>

        <div class="lesson-list" id="sprint-5-lessons" style="display: none;"><a href="/csp/sprint5/objectives"
                 class="lesson-item "
                 data-lesson-id="/csp/sprint5/objectives">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Team Project Objectives</span>
              </a></div>
      </div><div class="sprint-section">
        <button class="sprint-toggle" onclick="toggleSprint('6')" type="button">
          <i class="fas fa-chevron-right sprint-chevron" id="chevron-6"></i>
          <div class="sprint-info">
            <div class="sprint-title">Sprint6: Non-Profit Exploration and Ideation</div>
            <div class="sprint-weeks">Weeks 21-24</div>
          </div>
        </button>

        <div class="lesson-list" id="sprint-6-lessons" style="display: none;"></div>
      </div><div class="sprint-section">
        <button class="sprint-toggle" onclick="toggleSprint('7')" type="button">
          <i class="fas fa-chevron-right sprint-chevron" id="chevron-7"></i>
          <div class="sprint-info">
            <div class="sprint-title">Sprint7: AI, Data, and Emerging Technology</div>
            <div class="sprint-weeks">Weeks 25-28</div>
          </div>
        </button>

        <div class="lesson-list" id="sprint-7-lessons" style="display: none;"><a href="/pandas/intro1"
                 class="lesson-item "
                 data-lesson-id="/pandas/intro1">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Data Frames | Pandas | Intro 1</span>
              </a><a href="/ml/titanic"
                 class="lesson-item "
                 data-lesson-id="/ml/titanic">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">ML | Titanic Data</span>
              </a><a href="/2025/02/20/pandas-ml_fitness_IPYNB_2_.html"
                 class="lesson-item "
                 data-lesson-id="/2025/02/20/pandas-ml_fitness_IPYNB_2_.html">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">ML | Fitness</span>
              </a><a href="/2025/02/20/pandas-ml-neural_IPYNB_2_.html"
                 class="lesson-item "
                 data-lesson-id="/2025/02/20/pandas-ml-neural_IPYNB_2_.html">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">ML | Neural Network | Handwritting Detection</span>
              </a><a href="/2025/02/19/pandas-df_intro2_IPYNB_2_.html"
                 class="lesson-item "
                 data-lesson-id="/2025/02/19/pandas-df_intro2_IPYNB_2_.html">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Data Frames | Pandas | Intro 2</span>
              </a><a href="/network/stack"
                 class="lesson-item "
                 data-lesson-id="/network/stack">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Network Stack | HTTP and TCP/IP</span>
              </a><a href="/2025/01/13/request_response_IPYNB_2_.html"
                 class="lesson-item "
                 data-lesson-id="/2025/01/13/request_response_IPYNB_2_.html">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">API | Request | Response | Database</span>
              </a><a href="/2025/01/08/sqlconnect_IPYNB_2_.html"
                 class="lesson-item "
                 data-lesson-id="/2025/01/08/sqlconnect_IPYNB_2_.html">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Data | SQL Connect</span>
              </a><a href="/2025/01/08/sqlalchemy_IPYNB_2_.html"
                 class="lesson-item "
                 data-lesson-id="/2025/01/08/sqlalchemy_IPYNB_2_.html">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Data | SQLAlchemy</span>
              </a><a href="/2025/01/06/binary_logic_IPYNB_2_.html"
                 class="lesson-item "
                 data-lesson-id="/2025/01/06/binary_logic_IPYNB_2_.html">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Data | Binary Logic</span>
              </a><a href="/2024/12/02/python-flask_in_jupyter_IPYNB_2_.html"
                 class="lesson-item "
                 data-lesson-id="/2024/12/02/python-flask_in_jupyter_IPYNB_2_.html">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Computer System | Web Server | Flask</span>
              </a><a href="/1.4/errors"
                 class="lesson-item "
                 data-lesson-id="/1.4/errors">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Topic 1.4 - Identifying and Correcting Errors</span>
              </a><a href="/coding/srp/csp/"
                 class="lesson-item "
                 data-lesson-id="/coding/srp/csp/">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Single Responsibility & API Chaining</span>
              </a><a href="/devops/aws-docker-deployment/part2"
                 class="lesson-item "
                 data-lesson-id="/devops/aws-docker-deployment/part2">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Computing Systems | AWS Deployment | Setup Applicationa</span>
              </a><a href="/devops/aws-docker-deployment/part1"
                 class="lesson-item "
                 data-lesson-id="/devops/aws-docker-deployment/part1">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Computing Systems | AWS Deployment | Launch EC2</span>
              </a><a href="/aws/deployment"
                 class="lesson-item "
                 data-lesson-id="/aws/deployment">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Computing System | AWS Deployment| Step-by-Step Guide</span>
              </a></div>
      </div><div class="sprint-section">
        <button class="sprint-toggle" onclick="toggleSprint('8')" type="button">
          <i class="fas fa-chevron-right sprint-chevron" id="chevron-8"></i>
          <div class="sprint-info">
            <div class="sprint-title">Sprint8: AP Review and Non-Profit Project Planning</div>
            <div class="sprint-weeks">Weeks 29-32</div>
          </div>
        </button>

        <div class="lesson-list" id="sprint-8-lessons" style="display: none;"><a href="/csp/ap-csp-comprehensive/"
                 class="lesson-item "
                 data-lesson-id="/csp/ap-csp-comprehensive/">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">AP CSP Comprehensive Study Sheet</span>
              </a><a href="/csp/pseudocode-runner/"
                 class="lesson-item "
                 data-lesson-id="/csp/pseudocode-runner/">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">AP CSP Pseudocode Runner</span>
              </a></div>
      </div><div class="sprint-section">
        <button class="sprint-toggle" onclick="toggleSprint('9')" type="button">
          <i class="fas fa-chevron-right sprint-chevron" id="chevron-9"></i>
          <div class="sprint-info">
            <div class="sprint-title">Sprint9: Non-Profit Development and Delivery</div>
            <div class="sprint-weeks">Weeks 33-37</div>
          </div>
        </button>

        <div class="lesson-list" id="sprint-9-lessons" style="display: none;"><a href="/csp/sprint9/objectives"
                 class="lesson-item "
                 data-lesson-id="/csp/sprint9/objectives">
                <i class="fas fa-circle lesson-status-icon incomplete"></i>
                <span class="lesson-title">Sprint 9 - CSP Objectives (Final)</span>
              </a></div>
      </div></div>
</div>


  <!-- MAIN CONTENT -->
  <div class="lesson-main">
    <!-- Top Bar -->
    <!-- LESSON PLAYER TOP BAR - Navigation and progress -->


<div class="topbar-lesson">
  <div class="topbar-left">
    <div class="breadcrumbs">
      <a href="/navigation/courses/csp" title="Back to course">
        <i class="fas fa-home"></i>
      </a>
      <i class="fas fa-chevron-right"></i>
      <span>Week 2</span>
      <i class="fas fa-chevron-right"></i>
      <span class="current">Onboarding Challenge - Ground 0</span>
      <button id="topbar-complete-btn" onclick="toggleComplete()" type="button" title="Mark lesson complete">
        <i class="fas fa-circle" id="complete-icon-topbar"></i>
        <span id="complete-text-topbar">Burndown</span>
      </button>
    </div>
  </div>
  <div class="topbar-progress">
    <div class="topbar-progress-fill" id="lesson-progress"></div>
  </div>
</div>


    <!-- Lesson Content -->
    <div class="lesson-content">
      <div class="lesson-content-inner">
        <div class="lesson-header">
          <h1>Onboarding Challenge - Ground 0</h1>
          <div class="lesson-meta"><span><i class="fas fa-clock"></i> <span class="read-time" title="Estimated read time">
    
     18 min read  </span></span>
            <span>•</span>
            <span><i class="fas fa-file-alt"></i> Assignment</span>
            
          </div>
        </div>

        <div class="lesson-body"><ul><li><a href="#task-centered-introduction">Task-Centered Introduction</a><ul><li><a href="#the-challenge">The Challenge</a></li><li><a href="#the-problem">The Problem</a></li><li><a href="#the-goal">The Goal</a></li><li><a href="#why-this-matters">Why This Matters</a></li></ul></li><li><a href="#activation--you-already-know-more-than-you-think">Activation — You Already Know More Than You Think</a><ul><li><a href="#what-you-already-know">What You Already Know</a></li><li><a href="#your-hidden-advantage">Your Hidden Advantage</a></li><li><a href="#the-pbl-mindset">The PBL Mindset</a></li><li><a href="#your-mission">Your Mission</a></li></ul></li><li><a href="#demonstration--see-the-tools-and-workflow-in-action">Demonstration — See the Tools and Workflow in Action</a><ul><li><a href="#the-foundation">The Foundation</a></li><li><a href="#watch-the-workflow">Watch the Workflow</a></li><li><a href="#what-youll-see">What You’ll See</a></li><li><a href="#from-tools-to-software-development">From Tools to Software Development</a></li><li><a href="#development-practices">Development Practices</a></li></ul></li><li><a href="#onboarding-objectives---ground-0">Onboarding Objectives - Ground 0</a><ul><li><a href="#1-understand-the-challenge">1. Understand the Challenge</a></li><li><a href="#2-my-development-environment">2. My Development Environment</a></li><li><a href="#3-my-github-workflow">3. My GitHub Workflow</a></li><li><a href="#4-forming-the-team">4. Forming the Team</a><ul><li><a href="#before-forming-the-team">Before Forming the Team</a></li><li><a href="#establishing-the-team">Establishing the Team</a></li><li><a href="#beginning-to-operate-as-a-team">Beginning to Operate as a Team</a></li></ul></li><li><a href="#5-beginning-the-project">5. Beginning the Project</a></li></ul></li><li><a href="#ground-0-team-activities">Ground 0 Team Activities</a><ul><li><a href="#team-formation">Team Formation</a></li><li><a href="#team-initialization">Team Initialization</a></li><li><a href="#team-practice">Team Practice</a></li></ul></li><li><a href="#technical--learning-evaluation">Technical &amp; Learning Evaluation</a><ul><li><a href="#evaluation-matrix">Evaluation Matrix</a></li><li><a href="#tools--development-environment">Tools &amp; Development Environment</a></li><li><a href="#development--creation">Development &amp; Creation</a></li><li><a href="#technical-awareness--communication">Technical Awareness &amp; Communication</a></li><li><a href="#ranking-guidelines">Ranking Guidelines</a></li><li><a href="#before-the-evaluation">Before the Evaluation</a></li><li><a href="#peer-perspective">Peer Perspective</a></li><li><a href="#teacher-perspective">Teacher Perspective</a></li><li><a href="#compare-the-perspectives">Compare the Perspectives</a></li><li><a href="#retrospective-question">Retrospective Question</a></li><li><a href="#strongest-technical-evidence">Strongest Technical Evidence</a></li></ul></li><li><a href="#professional-evaluation">Professional Evaluation</a><ul><li><a href="#evaluation-matrix-1">Evaluation Matrix</a></li><li><a href="#core-behaviors">Core Behaviors</a></li><li><a href="#collaboration">Collaboration</a></li><li><a href="#professional-skills">Professional Skills</a></li><li><a href="#ranking-guidelines-1">Ranking Guidelines</a></li><li><a href="#before-the-evaluation-1">Before the Evaluation</a></li><li><a href="#peer-perspective-1">Peer Perspective</a></li><li><a href="#teacher-perspective-1">Teacher Perspective</a></li><li><a href="#compare-the-perspectives-1">Compare the Perspectives</a></li><li><a href="#retrospective-question-1">Retrospective Question</a></li><li><a href="#next-sprint-commitment">Next Sprint Commitment</a></li><li><a href="#review--becoming-proficient-in-scrum-agile-workflow">Review – “Becoming proficient in Scrum Agile Workflow”</a></li></ul></li></ul>
<h2 id="task-centered-introduction">Task-Centered Introduction</h2>

<p>Welcome to the <strong>Open Coding Society</strong> — where you’ll develop professional software engineering skills while building real projects with your classmates.</p>

<h3 id="the-challenge">The Challenge</h3>

<p>Starting a new development environment can be overwhelming. You will encounter unfamiliar tools, new workflows, new teammates, and problems that do not always have obvious answers.</p>

<p><strong>This is your first challenge: learn to work in an unfamiliar development environment.</strong></p>

<p>Instead of simply following a set of instructions, you will document your setup process, troubleshoot problems, make decisions, and create solutions that help both you and your teammates succeed.</p>

<p>You are beginning to work as a <strong>student developer</strong>. Your work will include:</p>

<ul>
  <li>
    <p><strong>Mastering professional development tools</strong></p>

    <ul>
      <li>VS Code</li>
      <li>GitHub</li>
      <li>Git</li>
      <li>Development and testing workflows</li>
    </ul>
  </li>
  <li>
    <p><strong>Building web-based projects</strong></p>

    <ul>
      <li>Individual GitHub Pages portfolio</li>
      <li>Team project sites</li>
      <li>Project documentatio</li>
    </ul>
  </li>
  <li>
    <p><strong>Developing collaboration practices</strong></p>

    <ul>
      <li>Issues</li>
      <li>Kanban boards</li>
      <li>GitHub workflows</li>
      <li>Team communication</li>
    </ul>
  </li>
  <li>
    <p><strong>Beginning your course-specific development</strong></p>

    <ul>
      <li><strong>CSSE:</strong> Begin ideation and development of a new game while learning the OCS game engine.</li>
      <li><strong>CSP:</strong> Begin ideation and development of an OCS project or contribute to and extend a nonprofit project.</li>
      <li><strong>CSA:</strong> Begin ideation, design, development, maintenance, and testing of an existing OCS system.</li>
      <li><strong>CSH:</strong> Begin research and development focused on extending OCS intelligence.</li>
    </ul>
  </li>
  <li>
    <p><strong>Documenting your learning journey</strong></p>

    <ul>
      <li>Setup instructions</li>
      <li>Troubleshooting notes</li>
      <li>Decisions and solutions</li>
      <li>Learning reflections</li>
    </ul>
  </li>
</ul>

<p>This class mirrors the experience of <strong>professional software engineering onboarding</strong>. You are not expected to know everything when you begin. You are expected to learn how to find your way through unfamiliar problems.</p>

<h3 id="the-problem">The Problem</h3>

<p>New developers often feel overwhelmed by unfamiliar tools, workflows, technologies, and expectations. A problem that seems simple can become difficult when you do not yet understand the environment in which you are working.</p>

<p>Your job is not simply to get everything working and perfect it.</p>

<p><strong>Your job is to learn how to get things working and keep improving.</strong></p>

<p>That means learning how to investigate a problem, ask useful questions, find information, test possible solutions, document what you discover, and help the next person who encounters the same problem.</p>

<h3 id="the-goal">The Goal</h3>

<p>By the end of the first onboarding sprint, you will:</p>

<ul>
  <li><strong>Build Your Personal Portfolio:</strong> Have your individual GitHub Pages site online and validate that your required code runners and development tools are working.</li>
  <li><strong>Build Working Proficiency:</strong> Become comfortable using the essential tools and workflows required for your course.</li>
  <li><strong>Establish Your Development Workflow:</strong> Understand how you will create, test, document, and submit your work.</li>
  <li><strong>Document Your Progress:</strong> Create useful issues, Kanban boards, plans, and technical notes that guide your development.</li>
  <li><strong>Begin Your Project:</strong> Move from onboarding into meaningful development work appropriate to your course.</li>
  <li><strong>Reflect on Your Learning:</strong> Begin identifying what you know, what you do not yet know, how you respond to problems, and how your approach changes as you gain experience.</li>
</ul>

<pre><code class="language-mermaid">mindmap
  root((Student Developer))
    🛠️ Develop Skills
      VS Code
      GitHub
      Git
      Development Workflow
    🏗️ Build Projects
      Team Collaboration
      Individual Portfolio
      GitHub Pages
      Course Project
    📝 Document Progress
      Setup Instructions
      Troubleshooting Notes
      Issues
      Plans
      Learning Reflections
</code></pre>

<p><strong>Professional software developers use these same tools and workflows when joining new teams, learning unfamiliar systems, and building applications.</strong></p>

<h3 id="why-this-matters">Why This Matters</h3>

<p>The skills you develop during onboarding form the foundation for your future coursework.</p>

<p>If you learn how to work effectively in an unfamiliar technical environment, you will spend less time fighting your tools and more time solving meaningful problems.</p>

<p>More importantly, you will begin developing an engineering habit that will serve you throughout the year:</p>

<blockquote>
  <p><strong>Encounter a problem → investigate → experiment → solve → document → reflect → improve.</strong></p>
</blockquote>

<p>This onboarding challenge is the beginning of that process.</p>

<h2 id="activation--you-already-know-more-than-you-think">Activation — You Already Know More Than You Think</h2>

<blockquote>
  <p><strong>You are a learner, a developer, and a teacher.</strong></p>

  <p>Every success, error, question, and breakthrough is something you can learn from — and something that may help someone else.</p>
</blockquote>

<p>You are not starting from zero.</p>

<p>You have already spent years learning how to learn, solve problems, work with others, and respond when something does not make sense.</p>

<h3 id="what-you-already-know">What You Already Know</h3>

<table>
  <thead>
    <tr>
      <th>You’ve Used</th>
      <th>Now You’ll Use It As a Developer</th>
    </tr>
  </thead>
  <tbody>
    <tr>
      <td><strong>Rubrics</strong></td>
      <td>Define what “good” looks like</td>
    </tr>
    <tr>
      <td><strong>Checklists</strong></td>
      <td>Build repeatable processes</td>
    </tr>
    <tr>
      <td><strong>Assessments</strong></td>
      <td>Test whether something works</td>
    </tr>
    <tr>
      <td><strong>Feedback</strong></td>
      <td>Improve your work</td>
    </tr>
    <tr>
      <td><strong>Teamwork</strong></td>
      <td>Collaborate on a real project</td>
    </tr>
    <tr>
      <td><strong>Creative Problem Solving</strong></td>
      <td>Find solutions when there is no answer key</td>
    </tr>
  </tbody>
</table>

<p>And you already know the basic progression of learning:</p>

<p><strong>Remember → Understand → Apply → Analyze → Evaluate → Create</strong></p>

<p>In software development, that might look like:</p>

<p><strong>Know a command → understand what it does → use it → troubleshoot it → choose between solutions → build something new</strong></p>

<p>That’s the transition from <strong>student</strong> to <strong>student developer</strong>.</p>

<h3 id="your-hidden-advantage">Your Hidden Advantage</h3>

<p>You’ve spent more than a decade being a student.</p>

<p>You’ve experienced:</p>

<ul>
  <li>Great instructions — and confusing ones</li>
  <li>Helpful feedback — and feedback that wasn’t</li>
  <li>Teams that worked — and teams that struggled</li>
  <li>Assignments that engaged you — and assignments that didn’t</li>
</ul>

<p><strong>You already know what it feels like to be the learner.</strong></p>

<p>Now you get to use that experience to become a better developer.</p>

<h3 id="the-pbl-mindset">The PBL Mindset</h3>

<p>Project-based learning changes the question.</p>

<p>Instead of:</p>

<blockquote>
  <p><strong>“What does the teacher want me to do?”</strong></p>
</blockquote>

<p>Start asking:</p>

<blockquote>
  <p><strong>“What do I need to figure out to move the project forward?”</strong></p>
</blockquote>

<p>You will:</p>

<p><strong>Encounter → Investigate → Experiment → Solve → Document → Reflect → Improve</strong></p>

<h3 id="your-mission">Your Mission</h3>

<p>You are the beginner.</p>

<p>That is an advantage.</p>

<p>When something is confusing, <strong>capture the confusion</strong>.</p>

<p>When something breaks, <strong>document the solution</strong>.</p>

<p>When you discover a better way, <strong>share it</strong>.</p>

<p>When your team struggles, <strong>look for a better process</strong>.</p>

<p>Your struggles are not evidence that you are failing.</p>

<p><strong>They are evidence that you are learning.</strong></p>

<p>And what you learn can become the documentation, tools, and knowledge that help the next student succeed.</p>

<h2 id="demonstration--see-the-tools-and-workflow-in-action">Demonstration — See the Tools and Workflow in Action</h2>

<blockquote>
  <p><strong>Follow along:</strong> Open your own VS Code workspace and GitHub repository as we work through this demonstration.</p>
</blockquote>

<p>So far, we’ve focused on <strong>how you think and learn</strong>.</p>

<p>Now let’s see the <strong>development environment</strong> you will use.</p>

<h3 id="the-foundation">The Foundation</h3>

<p>You will use these tools repeatedly throughout the year:</p>

<table>
  <thead>
    <tr>
      <th>Tool</th>
      <th>What It Does</th>
    </tr>
  </thead>
  <tbody>
    <tr>
      <td><strong>VS Code</strong></td>
      <td>Write and edit Markdown, JavaScript, and other project files</td>
    </tr>
    <tr>
      <td><strong>VS Code Marketplace</strong></td>
      <td>Find and install extensions that improve your development environment</td>
    </tr>
    <tr>
      <td><strong>Make</strong></td>
      <td>Automate repetitive development tasks</td>
    </tr>
    <tr>
      <td><strong>Git</strong></td>
      <td>Track changes and build a history of your work</td>
    </tr>
    <tr>
      <td><strong>GitHub</strong></td>
      <td>Store, share, and collaborate on your code</td>
    </tr>
    <tr>
      <td><strong>GitHub Actions</strong></td>
      <td>Automatically build, test, and publish your projects</td>
    </tr>
    <tr>
      <td><strong>GitHub Pages</strong></td>
      <td>Publish your web projects</td>
    </tr>
  </tbody>
</table>

<h3 id="watch-the-workflow">Watch the Workflow</h3>

<p>The important thing is not learning each tool in isolation.</p>

<p><strong>Watch how they work together.</strong></p>

<pre><code class="language-mermaid">flowchart LR
    A[VS Code&lt;br/&gt;Create &amp; Edit]
    B[Make&lt;br/&gt;Build &amp; Test]
    C[Git&lt;br/&gt;Commit Changes]
    D[GitHub&lt;br/&gt;Push &amp; Sync]
    E[GitHub Actions&lt;br/&gt;Automate]
    F[GitHub Pages&lt;br/&gt;Publish]

    A --&gt; B --&gt; C --&gt; D --&gt; E --&gt; F
</code></pre>

<h3 id="what-youll-see">What You’ll See</h3>

<p>During the demonstration, we will:</p>

<ol>
  <li><strong>Open VS Code</strong> and review the key extensions.</li>
  <li><strong>Explore this notebook</strong> and the Markdown used to create it.</li>
  <li><strong>Make a change</strong> to a file.</li>
  <li><strong>Build or test the change</strong> using Make.</li>
  <li><strong>Commit the change</strong> using Git through VS Code.</li>
  <li><strong>Push the commit to GitHub.</strong></li>
  <li><strong>Watch GitHub Actions</strong> run automatically.</li>
  <li><strong>Verify the published result</strong> on GitHub Pages.</li>
</ol>

<h3 id="from-tools-to-software-development">From Tools to Software Development</h3>

<p>These tools support the <strong>Software Development Lifecycle (SDLC)</strong>.</p>

<p>As you progress, the simple onboarding loop will grow into a larger engineering workflow:</p>

<pre><code class="language-mermaid">flowchart TD
    A[Plan&lt;br/&gt;Issues &amp; Backlog]
    B[Develop&lt;br/&gt;Code &amp; Build]
    C[Test&lt;br/&gt;Unit &amp; System Tests]
    D[Version&lt;br/&gt;Commit &amp; Push]
    E[Collaborate&lt;br/&gt;Review &amp; Merge]
    F[Deploy&lt;br/&gt;Publish]
    G[Reflect&lt;br/&gt;Issues &amp; Replan]

    A --&gt; B --&gt; C --&gt; D --&gt; E --&gt; F --&gt; G
    G --&gt; A
</code></pre>

<h3 id="development-practices">Development Practices</h3>

<p>The onboarding challenge is the entry point to a the set of development practices.</p>

<table>
  <thead>
    <tr>
      <th>Practice</th>
      <th>Purpose</th>
    </tr>
  </thead>
  <tbody>
    <tr>
      <td><strong>Tools &amp; Equipment</strong></td>
      <td>The development environment used to build, test, and deploy software.</td>
    </tr>
    <tr>
      <td><strong>GitHub Pages</strong></td>
      <td>The framework for a web project — pages, layouts, Markdown, HTML, JavaScript, and how everything connects.</td>
    </tr>
    <tr>
      <td><strong>Agile Methodologies</strong></td>
      <td>How teams organize work, make decisions, and select project teams. A key focus of Agile is to use short cycles and iterate often</td>
    </tr>
    <tr>
      <td><strong>Sprint Framework</strong></td>
      <td>How work is organized into short, focused development cycles with checkpoints.</td>
    </tr>
  </tbody>
</table>

<h2 id="onboarding-objectives---ground-0">Onboarding Objectives - Ground 0</h2>

<p><strong>Check your starting position.</strong></p>

<p>You have spent <strong>Weeks 0–1</strong> setting up your development environment, working through guides, solving problems, and beginning to explore ideas according to your skill level.</p>

<p>Now, as you enter <strong>Weeks 2–3</strong>, make sure you are ready to move your mind from:</p>

<blockquote>
  <p><strong>Onboarding → Project Development</strong></p>
</blockquote>

<table>
  <thead>
    <tr>
      <th>Section</th>
      <th>Ready?</th>
    </tr>
  </thead>
  <tbody>
    <tr>
      <td><a href="#1-understand-the-challenge"><strong>1. Understand the Challenge</strong></a></td>
      <td>☐</td>
    </tr>
    <tr>
      <td><a href="#2-my-development-environment"><strong>2. My Development Environment</strong></a></td>
      <td>☐</td>
    </tr>
    <tr>
      <td><a href="#3-my-github-workflow"><strong>3. My GitHub Workflow</strong></a></td>
      <td>☐</td>
    </tr>
    <tr>
      <td><a href="#4-forming-the-team"><strong>4. Forming the Team</strong></a></td>
      <td>☐</td>
    </tr>
    <tr>
      <td><a href="#5-beginning-the-project"><strong>5. Beginning the Project</strong></a></td>
      <td>☐</td>
    </tr>
  </tbody>
</table>

<h3 id="1-understand-the-challenge">1. Understand the Challenge</h3>

<ul class="task-list">
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I understand that the goal was not simply to install tools, but to learn how to work through an unfamiliar development environment.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I can identify at least one problem I encountered and explain how I solved or investigated it.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I have documented something that could help another student.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I have a story about how forming a team, positives and challenges.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I have a story about starting a team project, positives and challenges.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I can explain the <strong>Onboarding Objectives</strong> in my own words.</li>
</ul>

<hr />

<h3 id="2-my-development-environment">2. My Development Environment</h3>

<ul class="task-list">
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I can open my development environment and begin working without waiting for someone else.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I can edit and preview the files I will use in my course.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I can use Git/GitHub to commit and push my work.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I know where to find the tool setup and troubleshooting guides.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I know how to ask for help when I am stuck.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I have a working strategy for my personal computer and development environment.</li>
</ul>

<hr />

<h3 id="3-my-github-workflow">3. My GitHub Workflow</h3>

<ul class="task-list">
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I can locate my GitHub repository.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I can create an issues in my GitHub repository location.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I can commit a change in GitHub.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I can create or edit a file in VSCode.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I can commint and sync changes from VSCode to GitHub.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I know about GitHub Actions and can determine if my site is working.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I know how to load my GitHub Pages portfolio site in my browser.</li>
</ul>

<hr />

<h3 id="4-forming-the-team">4. Forming the Team</h3>

<h4 id="before-forming-the-team">Before Forming the Team</h4>

<ul class="task-list">
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I have completed or reviewed the team formation process.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I have read an article on selecting effective development teams.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I have identified my <strong>persona, skills, interests, and strengths</strong> in OCS.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I understand the expectation for diversity of skills, interests, and perspectives when forming teams.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I have considered which teammates I may work well with.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I have sat in proximity to potential teammates.</li>
</ul>

<h4 id="establishing-the-team">Establishing the Team</h4>

<ul class="task-list">
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I know my team members’ names.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />We have discussed our skills, interests, and perspectives.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />We have established a team communication channel.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I have shared my preferred contact method with my teammates.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />We have established a team <strong>Agile manifesto and communication protocols</strong>.</li>
</ul>

<h4 id="beginning-to-operate-as-a-team">Beginning to Operate as a Team</h4>

<ul class="task-list">
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />We have identified the coding challenges required for our course.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />We have created team and individual GitHub repositories.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />We have created team and individual GitHub issues, kanban, and milestones.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />We have begun team decision-making processes such as stand-ups, pin-ups, and burndowns.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />We have practiced Live Share, debugging, or pair programming.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />We have established a process for discussing problems and resolving issues.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />We have established our team contribution workflows: fork, branch, pull request, or direct collaboration as appropriate.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />We have identified how we will accept feedback and take action between checkpoints.</li>
</ul>

<hr />

<h3 id="5-beginning-the-project">5. Beginning the Project</h3>

<ul class="task-list">
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I have participated in project ideation.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I can describe at least one problem or project idea our team is considering.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I have contributed an idea, question, research finding, or technical consideration.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I have created or contributed to a GitHub issue related to our work.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />We have established our first project direction.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I understand what tasks I am expected to accomplish before the next sprint checkpoint.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />We have identified what the team needs to accomplish before the next checkpoint.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I know that presenting evidence is part of my responsibility as I work.</li>
</ul>

<hr />

<h2 id="ground-0-team-activities">Ground 0 Team Activities</h2>

<h3 id="team-formation">Team Formation</h3>

<p><strong>Understand → Introduce → Explore → Form</strong></p>

<p>Read, discuss, and use the team-formation process to create groups based on <strong>skills, interests, perspectives, and project needs</strong>, rather than friendship alone.</p>

<h3 id="team-initialization">Team Initialization</h3>

<p><strong>Organize → Connect → Decide → Build</strong></p>

<p>Establish the team’s communication, GitHub workflow, coding practices, decision-making processes, and initial project direction.</p>

<h3 id="team-practice">Team Practice</h3>

<p><strong>Communicate → Collaborate → Solve → Document</strong></p>

<p>Begin practicing the behaviors expected throughout the sprint:</p>

<ul>
  <li>Daily checkpoints</li>
  <li>Issue tracking</li>
  <li>Pair programming</li>
  <li>Live debugging</li>
  <li>Peer support</li>
  <li>Decision documentation</li>
  <li>Evidence collection</li>
  <li>Integration and Test policies</li>
  <li>Preparing for Checkpoints and Demos</li>
</ul>

<h2 id="technical--learning-evaluation">Technical &amp; Learning Evaluation</h2>

<p>This checklist looks beyond whether you <strong>completed a task</strong>.</p>

<p>Technical development includes <strong>using tools, building solutions, solving problems, learning from mistakes, and becoming more independent</strong>.</p>

<p>Your evaluation should show not only <strong>what you completed</strong>, but how your technical ability changed during the sprint.</p>

<h3 id="evaluation-matrix">Evaluation Matrix</h3>

<p>Supply a GitHub Issue with your <strong>self-evaluation scores and one summary for each section</strong>.</p>

<blockquote>
  <p><strong>Rank each area individually. Write one reflection for each section.</strong></p>

  <p>You do not need to write a separate response for every row.</p>
</blockquote>

<h3 id="tools--development-environment">Tools &amp; Development Environment</h3>

<table>
  <thead>
    <tr>
      <th>ID</th>
      <th>Area</th>
      <th style="text-align: right">Self Rank</th>
      <th style="text-align: center">Peer</th>
      <th style="text-align: center">Teacher</th>
    </tr>
  </thead>
  <tbody>
    <tr>
      <td>T01</td>
      <td>Development Environment</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
    <tr>
      <td>T02</td>
      <td>VS Code</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
    <tr>
      <td>T03</td>
      <td>Git / GitHub</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
    <tr>
      <td>T04</td>
      <td>Portfolio / GitHub Pages</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
  </tbody>
</table>

<p><strong>Summary:</strong><br />
Describe how your ability to use your development environment changed during the sprint. Include a specific setup problem, tool, workflow, or technical process that you learned to manage.</p>

<blockquote>

</blockquote>

<hr />

<h3 id="development--creation">Development &amp; Creation</h3>

<table>
  <thead>
    <tr>
      <th>ID</th>
      <th>Area</th>
      <th style="text-align: right">Self Rank</th>
      <th style="text-align: center">Peer</th>
      <th style="text-align: center">Teacher</th>
    </tr>
  </thead>
  <tbody>
    <tr>
      <td>T05</td>
      <td>Tools &amp; Equipment Hacks</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
    <tr>
      <td>T06</td>
      <td>Portfolio / Blogging Hacks</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
    <tr>
      <td>T07</td>
      <td>Theme / Style / Layout</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
    <tr>
      <td>T08</td>
      <td>JavaScript / Coding Challenges</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
    <tr>
      <td>T09</td>
      <td>AI Orchestration</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
  </tbody>
</table>

<p><strong>Summary:</strong><br />
Describe something you built, modified, customized, or extended during the sprint. Explain a tool, technique, coding challenge, design decision, or <strong>AI orchestration process</strong> that helped you create or improve your work.</p>

<blockquote>

</blockquote>

<hr />

<h3 id="technical-awareness--communication">Technical Awareness &amp; Communication</h3>

<table>
  <thead>
    <tr>
      <th>ID</th>
      <th>Area</th>
      <th style="text-align: right">Self Rank</th>
      <th style="text-align: center">Peer</th>
      <th style="text-align: center">Teacher</th>
    </tr>
  </thead>
  <tbody>
    <tr>
      <td>T10</td>
      <td>Tech / Cyber Growth</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
    <tr>
      <td>T11</td>
      <td>Learning Through Mistakes</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
    <tr>
      <td>T12</td>
      <td>Tech / Cyber Talk</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
  </tbody>
</table>

<p><strong>Summary:</strong><br />
Describe how your technical understanding and communication developed during the sprint. Include a specific example of something you learned, a mistake that changed your understanding, or a technical idea you can now explain more clearly.</p>

<blockquote>

</blockquote>

<hr />

<h3 id="ranking-guidelines">Ranking Guidelines</h3>

<p>Rank each area based on the <strong>technical ability you demonstrated during the sprint</strong>.</p>

<blockquote>
  <p><strong>0 means you have not entered a rank yet.</strong></p>

  <p>Start with 1 and select the level that best represents your demonstrated ability.</p>
</blockquote>

<table>
  <thead>
    <tr>
      <th style="text-align: right">Rank</th>
      <th>Description</th>
    </tr>
  </thead>
  <tbody>
    <tr>
      <td style="text-align: right"><strong>0</strong></td>
      <td><strong>Not Entered</strong> — No self-assessment has been entered yet.</td>
    </tr>
    <tr>
      <td style="text-align: right"><strong>1</strong></td>
      <td><strong>Explore</strong> — I have encountered or experimented with this technology or skill.</td>
    </tr>
    <tr>
      <td style="text-align: right"><strong>2</strong></td>
      <td><strong>Understand</strong> — I can explain it and perform basic tasks with support.</td>
    </tr>
    <tr>
      <td style="text-align: right"><strong>3</strong></td>
      <td><strong>Apply</strong> — I can use it independently in my project work.</td>
    </tr>
    <tr>
      <td style="text-align: right"><strong>4</strong></td>
      <td><strong>Analyze / Evaluate</strong> — I can troubleshoot problems, compare approaches, and improve my solution.</td>
    </tr>
    <tr>
      <td style="text-align: right"><strong>5</strong></td>
      <td><strong>Create / Extend</strong> — I can independently customize, extend, or create solutions using the skill.</td>
    </tr>
  </tbody>
</table>

<p><strong>Peer / Teacher Calibration:</strong> <code class="language-plaintext highlighter-rouge">−</code> = below self-assessment · <code class="language-plaintext highlighter-rouge">=</code> = consistent with self-assessment · <code class="language-plaintext highlighter-rouge">+</code> = above self-assessment</p>

<hr />

<h3 id="before-the-evaluation">Before the Evaluation</h3>

<ul class="task-list">
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I reviewed my <strong>Ground 0 Checklist</strong>.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I reviewed the technical evidence collected during the sprint.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I completed my self-ranking before receiving peer or teacher feedback.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I can support my rankings with examples from my work.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I identified at least one technical area where I want to improve.</li>
</ul>

<h3 id="peer-perspective">Peer Perspective</h3>

<p>With your team:</p>

<ul class="task-list">
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I reviewed technical work or evidence from a teammate.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I identified a technical strength a teammate demonstrated.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I identified a constructive opportunity for a teammate to improve.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />My feedback is based on technical evidence or observed work rather than personal preference.</li>
</ul>

<h3 id="teacher-perspective">Teacher Perspective</h3>

<p>The teacher will use <strong>working code, repositories, commits, issues, demonstrations, documentation, troubleshooting, and discussions</strong> to evaluate technical growth.</p>

<h3 id="compare-the-perspectives">Compare the Perspectives</h3>

<p>After the peer and teacher evaluations are complete:</p>

<ul class="task-list">
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I compared my self-ranking with my peer ranking.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I compared my self-ranking with my teacher ranking.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I identified an area where the perspectives agreed.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I identified an area where the perspectives differed.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I considered why the perspectives may have differed.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I identified one specific technical action for my next sprint.</li>
</ul>

<h3 id="retrospective-question">Retrospective Question</h3>

<blockquote>
  <p><strong>What can I do now as a developer that I could not do at the beginning of this sprint?</strong></p>
</blockquote>

<h3 id="strongest-technical-evidence">Strongest Technical Evidence</h3>

<p><strong>The technical evidence I am most proud of:</strong></p>

<blockquote>
  <p>TBD</p>
</blockquote>

<p><strong>The biggest technical problem I solved:</strong></p>

<blockquote>
  <p>TBD</p>
</blockquote>

<p><strong>The technical skill I most need to improve:</strong></p>

<blockquote>
  <p>TBD</p>
</blockquote>

<p><strong>My next technical learning goal:</strong></p>

<blockquote>
  <p>TBD</p>
</blockquote>

<h2 id="professional-evaluation">Professional Evaluation</h2>

<p>This checklist and its summaries look beyond technical tasks.</p>

<p>Professional development includes <strong>how you work, communicate, collaborate, respond to problems, and contribute to a team</strong>.</p>

<p>Your performance will be considered from three perspectives:</p>

<p><strong>Self → Peer → Teacher</strong></p>

<h3 id="evaluation-matrix-1">Evaluation Matrix</h3>

<p>Supply a GitHub Issue with your <strong>self-evaluation scores and one summary for each section</strong>.</p>

<blockquote>
  <p><strong>Rank each area individually. Write one reflection for each section.</strong></p>

  <p>You do not need to write a separate response for every row.</p>
</blockquote>

<h3 id="core-behaviors">Core Behaviors</h3>

<table>
  <thead>
    <tr>
      <th>ID</th>
      <th>Area</th>
      <th style="text-align: right">Self Rank</th>
      <th style="text-align: center">Peer</th>
      <th style="text-align: center">Teacher</th>
    </tr>
  </thead>
  <tbody>
    <tr>
      <td>P01</td>
      <td>Attendance / Tardy</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
    <tr>
      <td>P02</td>
      <td>Work Habits</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
    <tr>
      <td>P03</td>
      <td>Integrity</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
  </tbody>
</table>

<p><strong>Summary:</strong><br />
Describe how you used your class time, responded to expectations, and demonstrated responsibility during the sprint. Include a specific example of a success, difficulty, or change you made.</p>

<blockquote>

</blockquote>

<hr />

<h3 id="collaboration">Collaboration</h3>

<table>
  <thead>
    <tr>
      <th>ID</th>
      <th>Area</th>
      <th style="text-align: right">Self Rank</th>
      <th style="text-align: center">Peer</th>
      <th style="text-align: center">Teacher</th>
    </tr>
  </thead>
  <tbody>
    <tr>
      <td>P04</td>
      <td>Communication</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
    <tr>
      <td>P05</td>
      <td>Help Seeking</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
    <tr>
      <td>P06</td>
      <td>Mentoring / Advocacy</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
  </tbody>
</table>

<p><strong>Summary:</strong><br />
Describe how you communicated, sought help, helped others, and participated in your team. Include a specific interaction or example.</p>

<blockquote>

</blockquote>

<hr />

<h3 id="professional-skills">Professional Skills</h3>

<table>
  <thead>
    <tr>
      <th>ID</th>
      <th>Area</th>
      <th style="text-align: right">Self Rank</th>
      <th style="text-align: center">Peer</th>
      <th style="text-align: center">Teacher</th>
    </tr>
  </thead>
  <tbody>
    <tr>
      <td>P07</td>
      <td>Timeliness</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
    <tr>
      <td>P08</td>
      <td>Persistence</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
    <tr>
      <td>P09</td>
      <td>Organization</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
    <tr>
      <td>P10</td>
      <td>Engagement</td>
      <td style="text-align: right">0</td>
      <td style="text-align: center">—</td>
      <td style="text-align: center">—</td>
    </tr>
  </tbody>
</table>

<p><strong>Summary:</strong><br />
Describe how you managed your time, persisted through difficulty, stayed organized, and remained engaged. Include a specific example of how you responded when the work became difficult or frustrating.</p>

<blockquote>

</blockquote>

<hr />

<h3 id="ranking-guidelines-1">Ranking Guidelines</h3>

<p>Rank each area based on the <strong>behavior you demonstrated during the sprint</strong>.</p>

<blockquote>
  <p><strong>0 means you have not entered a rank yet.</strong></p>

  <p>Start with 1 and select the level that best represents your demonstrated behavior.</p>
</blockquote>

<table>
  <thead>
    <tr>
      <th style="text-align: right">Rank</th>
      <th>Description</th>
    </tr>
  </thead>
  <tbody>
    <tr>
      <td style="text-align: right"><strong>0</strong></td>
      <td><strong>Not Entered</strong> — No self-assessment has been entered yet.</td>
    </tr>
    <tr>
      <td style="text-align: right"><strong>1</strong></td>
      <td><strong>Beginning</strong> — I am beginning to recognize and practice this behavior.</td>
    </tr>
    <tr>
      <td style="text-align: right"><strong>2</strong></td>
      <td><strong>Developing</strong> — I demonstrate this behavior sometimes, usually with reminders or support.</td>
    </tr>
    <tr>
      <td style="text-align: right"><strong>3</strong></td>
      <td><strong>Consistent</strong> — I regularly demonstrate this behavior independently.</td>
    </tr>
    <tr>
      <td style="text-align: right"><strong>4</strong></td>
      <td><strong>Strong</strong> — I consistently demonstrate this behavior and adjust when circumstances change.</td>
    </tr>
    <tr>
      <td style="text-align: right"><strong>5</strong></td>
      <td><strong>Model</strong> — I consistently demonstrate this behavior and help others develop it.</td>
    </tr>
  </tbody>
</table>

<p><strong>Peer / Teacher Calibration:</strong> <code class="language-plaintext highlighter-rouge">−</code> = below self-assessment · <code class="language-plaintext highlighter-rouge">=</code> = consistent with self-assessment · <code class="language-plaintext highlighter-rouge">+</code> = above self-assessment</p>

<hr />

<h3 id="before-the-evaluation-1">Before the Evaluation</h3>

<ul class="task-list">
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I reviewed my <strong>Ground 0 Checklist</strong>.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I reviewed the evidence collected during the sprint.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I completed my self-ranking before receiving peer or teacher feedback.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I can support my rankings with examples from my sprint.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I identified at least one area where I want to improve.</li>
</ul>

<h3 id="peer-perspective-1">Peer Perspective</h3>

<p>With your team:</p>

<ul class="task-list">
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I provided useful feedback to a teammate.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I identified a strength that a teammate contributed to the project.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I identified a constructive opportunity for a teammate to improve.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />My feedback is based on observed behavior or evidence rather than personal preference.</li>
</ul>

<h3 id="teacher-perspective-1">Teacher Perspective</h3>

<p>The teacher will use <strong>project evidence, observations, discussions, documentation, and team interactions</strong> to evaluate growth.</p>

<h3 id="compare-the-perspectives-1">Compare the Perspectives</h3>

<p>After the peer and teacher evaluations are complete:</p>

<ul class="task-list">
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I compared my self-ranking with my peer ranking.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I compared my self-ranking with my teacher ranking.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I identified an area where the perspectives agreed.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I identified an area where the perspectives differed.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I considered why the perspectives may have differed.</li>
  <li class="task-list-item"><input type="checkbox" class="task-list-item-checkbox" disabled="disabled" />I identified one specific action for my next sprint.</li>
</ul>

<h3 id="retrospective-question-1">Retrospective Question</h3>

<blockquote>
  <p><strong>What did I learn about myself as a developer and teammate that I could not have learned from a grade alone?</strong></p>
</blockquote>

<h3 id="next-sprint-commitment">Next Sprint Commitment</h3>

<p><strong>One professional behavior I will improve:</strong></p>

<blockquote>
  <p>TBD</p>
</blockquote>

<p><strong>One technical skill I will improve:</strong></p>

<blockquote>
  <p>TBD</p>
</blockquote>

<p><strong>One way I will contribute more effectively to my team:</strong></p>

<blockquote>
  <p>TBD</p>
</blockquote>

<p><strong>Evidence I will collect during the next sprint:</strong></p>

<blockquote>
  <p>TBD</p>
</blockquote>

<h3 id="review--becoming-proficient-in-scrum-agile-workflow">Review – “Becoming proficient in Scrum Agile Workflow”</h3>

<p>At the end of the sprint, teams consolidate their growth and show how they’ve applied <strong>technical, collaborative, and design skills</strong>.</p>

<pre><code class="language-mermaid">flowchart TD
    subgraph SR ["Sprint Review Cycle"]
        A[Team Capability Review] --&gt; B[Peer to Peer Presentation]
        B --&gt; C[Receive Feedback]
        C --&gt; D[Refinement &amp; Iteration]
        D --&gt; E[Teacher Close-Out]
        E --&gt; F[Individual Mastery Review]
        F --&gt; G[Set Goals for Next Sprint]
        G --&gt; A
    end
    
    F --&gt; H[Personal Growth]
    
</code></pre>


          

          
          
          <script>
  window.PageFeatures = window.PageFeatures || {};
  window.PageFeatures.grading_modules = true;
</script>

<!-- ═══════════════════════════════════════════════════════════════════
     PIXEL SUBMIT WIDGET
════════════════════════════════════════════════════════════════════ -->
<div class="grading-submission-module">
  <link rel="preconnect" href="https://fonts.googleapis.com">
<link href="https://fonts.googleapis.com/css2?family=Press+Start+2P&display=swap" rel="stylesheet">

<div id="pixel-submit-widget"
  data-gist-description="Exported AP CSA FRQs / Challenges from Open Coding Society"
>

  <!-- ── Title bar ── -->
  <div class="psw-titlebar">
    <span class="psw-titlebar-text">HOMEWORK DELIVERY</span>
  </div>

  <!-- ── Two panels ── -->
  <div class="psw-panels">

    <div class="psw-panel psw-panel-left">
      <div class="psw-step-label">STEP 1</div>
      <canvas id="psw-scroll-canvas" width="138" height="96"></canvas>
      <button id="psw-export-btn" class="psw-btn psw-btn-gold">✦ EXPORT</button>
    </div>

    <div class="psw-panel psw-panel-right">
      <div class="psw-step-label">STEP 2</div>
      <canvas id="psw-carrier-canvas" width="180" height="138"></canvas>
      <button id="psw-commit-btn" class="psw-btn psw-btn-blue" disabled>✉ COMMIT</button>
    </div>

  </div>

  <!-- ── Dialog box ── -->
  <div class="psw-dialog">
    <span id="psw-dialog-icon" class="psw-dialog-icon">✉</span>
    <span id="psw-dialog-text" class="psw-dialog-text psw-dialog-muted">
      Export your solutions, then commit to submit...
    </span>
  </div>

</div>

<script type="module" src="/assets/js/gist-export-button.js"></script>

</div>

<!-- Hidden admin button (shown by JS for admins only) -->
<button id="open-grades-admin" class="btn-secondary" style="display:none;margin:0 auto 1rem;">
  Grades Admin
</button>

<!-- ═══════════════════════════════════════════════════════════════════
     ADMIN PANEL (hidden until admin clicks)
════════════════════════════════════════════════════════════════════ -->
<div id="grading-module-widget">
  <div id="grading-module" class="grading-module-panel" style="display:none;">
    <h2>Grades Admin</h2>
    <section>
      <form id="grade-create-form">
        <input name="uid"                                   placeholder="Student UID"                          required />
        <input name="assignment"                            placeholder="Assignment"                           required />
        <input type="number" name="score" step="0.01"      placeholder="Score"                                required />
        <input name="teacherComments"                       placeholder="Teacher comments" />
        <input name="submission"                            placeholder="Submission (text, URL, comments)" />
        <button type="submit" class="btn-primary">Submit Grade</button>
      </form>
      <table id="grades-table" class="grading-table">
        <thead>
          <tr>
            <th>ID</th><th>UID</th><th>Assignment</th><th>Score</th>
            <th>Comments</th><th>Submission</th><th>Actions</th>
          </tr>
        </thead>
        <tbody></tbody>
      </table>
      <form id="grade-edit-form" style="display:none;margin-top:1rem;">
        <input type="hidden" name="id" />
        <input name="uid"              readonly />
        <input name="assignment"       required />
        <input type="number" name="score" step="0.01" required />
        <input name="teacherComments" />
        <input name="submission" />
        <button type="submit" class="btn-primary">Update</button>
        <button type="button" id="grade-edit-cancel" class="btn-secondary">Cancel</button>
      </form>
    </section>
    <div id="grading-status"></div>
  </div>
</div>

<!-- ═══════════════════════════════════════════════════════════════════
     MANUAL FALLBACK MODAL
     Opened automatically by the pixel widget if the API is unreachable
════════════════════════════════════════════════════════════════════ -->
<div id="grade-manual-modal" class="grading-modal-backdrop">
  <div class="grading-modal-box">
    <h3>Submit Failed — Enter Manually</h3>
    <p class="grading-modal-error" id="manual-error-msg"></p>
    <label>UID            <input id="manual-uid"        placeholder="Your UID / username" /></label>
    <label>Assignment     <input id="manual-assignment" placeholder="e.g. homework1.html" /></label>
    <label>Submission URL <input id="manual-submission" placeholder="https://yoursite.com/page" /></label>
    <label>Comments       <input id="manual-comments"   placeholder="Optional notes" /></label>
    <div class="grading-modal-actions">
      <button id="manual-cancel-btn" class="btn-secondary">Cancel</button>
      <button id="manual-send-btn"   class="btn-primary">Submit</button>
    </div>
  </div>
</div>

<script src="/assets/js/grading-module.js" defer></script>
          

          
          <!-- ASSIGNMENT SUBMISSION FORM -->


<div class="assignment-submission">
  <h3>
    <i class="fas fa-upload assignment-submission-icon"></i>
    Submit Assignment
  </h3>

  <div class="submission-options">
    <button class="submission-tab" onclick="switchSubmissionType('link')" type="button">
      <i class="fas fa-link"></i> Link Submission
    </button>
    <button class="submission-tab" onclick="switchSubmissionType('file')" type="button">
      <i class="fas fa-file-upload"></i> File Upload
    </button>
  </div>

  <!-- Link Submission Form -->
  <form class="submission-form" id="link-form">
    <div class="form-group">
      <label for="submission-link">Submission Link</label>
      <input type="url" id="submission-link" placeholder="https://..." required>
    </div>
    <div class="form-group">
      <label for="link-description">Description</label>
      <textarea id="link-description" placeholder="Describe what you're submitting..." required></textarea>
    </div>
    <button type="submit" class="submit-btn">
      <i class="fas fa-check"></i>
      Submit Link
    </button>
  </form>

  <!-- File Upload Form -->
  <form class="submission-form hidden" id="file-form">
    <div class="form-group">
      <label>Upload Files</label>
      <div class="file-upload-area" id="file-upload-area">
        <input type="file" id="file-input" multiple style="display: none;">
        <div class="file-upload-icon">
          <i class="fas fa-cloud-upload-alt"></i>
        </div>
        <div class="file-upload-text">Click to upload or drag and drop</div>
        <div class="file-upload-hint">PDF, ZIP, images, or documents (Max 10MB per file)</div>
      </div>
      <div class="file-list" id="file-list"></div>
    </div>
    <div class="form-group">
      <label for="file-notes">Notes (Optional)</label>
      <textarea id="file-notes" placeholder="Add any additional comments..."></textarea>
    </div>
    <button type="submit" class="submit-btn" id="file-submit-btn" disabled>
      <i class="fas fa-check"></i>
      Submit Files
    </button>
  </form>

  <div class="submission-status" id="submission-status"></div>
</div>

          <p style="margin-top: 12px; color: #9ca3af; font-size: 0.95rem;">
            Need to update a submission later? Open the <a href="/submissions" style="color: #60a5fa; text-decoration: underline;">submissions dashboard</a>.
          </p>
          

          
          <!-- Lesson Chat: collapsible WebSocket chat at the bottom of a lesson page. -->
<!-- Opt-in via `chat: true` in lesson frontmatter. -->
<!-- Reuses the STOMP/SockJS backend from group_dashboard.html. Per-lesson scope -->
<!-- is achieved by tagging messages with a [[lesson:<url>]] marker and filtering -->
<!-- on the client. All lessons share a single backbone group named "lessons". -->

<details id="lessonChat" data-lesson-key="/sprint1/challenge/csp/" style="margin-top:2em">
  <summary>Lesson Chat — <span id="lessonChatStatus">click to open</span></summary>
  <div id="lessonChatMessages" aria-live="polite" style="height:18em;overflow-y:auto;border:1px solid;padding:.5em;margin:.5em 0">
    <p id="lessonChatEmpty"><em>No messages yet.</em></p>
  </div>
  <form id="lessonChatForm" autocomplete="off" style="display:flex;gap:.5em">
    <input id="lessonChatInput" type="text" placeholder="Message this lesson…" maxlength="2000" disabled style="flex:1">
    <button id="lessonChatSend" type="submit" disabled>Send</button>
  </form>
</details>

<script src="https://cdnjs.cloudflare.com/ajax/libs/sockjs-client/1.5.1/sockjs.min.js"></script>
<script src="https://cdnjs.cloudflare.com/ajax/libs/stomp.js/2.3.3/stomp.min.js"></script>

<script type="module">
  import { javaURI, pythonURI, fetchOptions } from '/assets/js/api/config.js';

  const LESSON_KEY = "/sprint1/challenge/csp/";
  const LESSON_MARKER = `[[lesson:${LESSON_KEY}]]`;
  const LESSON_MARKER_RE = /^\s*\[\[lesson:([^\]]+)\]\]\s?/;
  const CHAT_SOCKET_PORT = 8589;

  function buildChatSocketEndpoint() {
    const uri = new URL(javaURI);
    if (uri.hostname === 'localhost' || uri.hostname === '127.0.0.1') {
      return `${uri.protocol}//${uri.hostname}:${CHAT_SOCKET_PORT}/ws-chat`;
    }
    return javaURI + '/ws-chat';
  }

  const state = {
    backboneGroupId: null,
    stompClient: null,
    subscription: null,
    connected: false,
    initialized: false,
    currentUserDisplayName: 'You',
    seenKeys: new Set(),
  };

  const detailsEl = document.getElementById('lessonChat');
  const statusEl = document.getElementById('lessonChatStatus');
  const messagesEl = document.getElementById('lessonChatMessages');
  const inputEl = document.getElementById('lessonChatInput');
  const sendBtn = document.getElementById('lessonChatSend');
  const formEl = document.getElementById('lessonChatForm');

  function setStatus(text) {
    statusEl.textContent = text;
  }

  function formatTimestamp(isoString) {
    if (!isoString) return '';
    const date = new Date(isoString);
    if (Number.isNaN(date.getTime())) return '';
    return date.toLocaleString(undefined, { month: 'short', day: 'numeric', hour: 'numeric', minute: '2-digit' });
  }

  function clearEmpty() {
    const empty = document.getElementById('lessonChatEmpty');
    if (empty) empty.remove();
  }

  function appendMessage({ sender, message, date }) {
    clearEmpty();
    const who = sender === state.currentUserDisplayName ? 'You' : (sender || 'Unknown');
    const when = formatTimestamp(date);
    const line = document.createElement('p');
    const strong = document.createElement('strong');
    strong.textContent = who;
    line.append(strong, document.createTextNode(when ? ` · ${when}: ` : ': '), document.createTextNode(message));
    messagesEl.appendChild(line);
    messagesEl.scrollTop = messagesEl.scrollHeight;
  }

  function appendSystem(text) {
    clearEmpty();
    const el = document.createElement('p');
    const em = document.createElement('em');
    em.textContent = text;
    el.appendChild(em);
    messagesEl.appendChild(el);
    messagesEl.scrollTop = messagesEl.scrollHeight;
  }

  function eventKey(event) {
    return [event?.sender || '', event?.date || '', event?.message || ''].join('|');
  }

  function matchesLesson(rawMessage) {
    const m = LESSON_MARKER_RE.exec(rawMessage || '');
    if (!m) return null;
    if (m[1] !== LESSON_KEY) return null;
    return (rawMessage || '').replace(LESSON_MARKER_RE, '');
  }

  async function resolveCurrentUser() {
    try {
      const res = await fetch(`${pythonURI}/api/id`, fetchOptions);
      if (!res.ok) {
        console.warn('Lesson chat: /api/id HTTP', res.status);
        return;
      }
      const user = await res.json();
      state.currentUserDisplayName = user?.name || user?.uid || user?.github || 'You';
    } catch (err) {
      console.warn('Lesson chat: user lookup failed', err);
    }
  }

  async function resolveBackboneGroup() {
    if (window.__lessonChatGroupId) {
      state.backboneGroupId = window.__lessonChatGroupId;
      return true;
    }
    const url = `${javaURI}/api/groups/search?name=lessons`;
    let res;
    try {
      res = await fetch(url, fetchOptions);
    } catch (err) {
      console.warn('Lesson chat: group lookup network error', err);
      return false;
    }
    if (!res.ok) {
      const text = await res.text().catch(() => '');
      console.warn('Lesson chat: group lookup HTTP', res.status, text);
      return false;
    }
    let body;
    try {
      body = await res.json();
    } catch (err) {
      console.warn('Lesson chat: group lookup non-JSON response', err);
      return false;
    }
    const match = Array.isArray(body)
      ? (body.find((g) => (g?.name || '').toLowerCase() === 'lessons') || body[0])
      : body;
    const id = Number(match?.id);
    if (!id) {
      console.warn('Lesson chat: no group named "lessons" in response', body);
      return false;
    }
    state.backboneGroupId = id;
    window.__lessonChatGroupId = id;
    return true;
  }

  async function loadHistory() {
    if (!state.backboneGroupId) return;
    try {
      const res = await fetch(`${javaURI}/api/groups/chat/${state.backboneGroupId}/messages`, fetchOptions);
      if (!res.ok) {
        console.warn('Lesson chat: history HTTP', res.status);
        return;
      }
      const messages = await res.json();
      (messages || []).forEach((msg) => {
        const stripped = matchesLesson(msg.message);
        if (stripped === null) return;
        const key = eventKey({ sender: msg.name, date: msg.date, message: msg.message });
        if (state.seenKeys.has(key)) return;
        state.seenKeys.add(key);
        appendMessage({ sender: msg.name, message: stripped, date: msg.date });
      });
    } catch (err) {
      console.warn('Lesson chat: history load failed', err);
    }
  }

  function handleIncoming(event) {
    if (!event || event.context !== 'sendMessageServer') return;
    const stripped = matchesLesson(event.message);
    if (stripped === null) return;
    const key = eventKey(event);
    if (state.seenKeys.has(key)) return;
    state.seenKeys.add(key);
    appendMessage({ sender: event.sender, message: stripped, date: event.date });
  }

  function connect() {
    return new Promise((resolve) => {
      if (state.connected) return resolve(true);
      if (typeof SockJS === 'undefined' || typeof Stomp === 'undefined') {
        setStatus('unavailable');
        return resolve(false);
      }
      setStatus('connecting…');
      const socket = new SockJS(buildChatSocketEndpoint());
      const client = Stomp.over(socket);
      client.debug = null;
      state.stompClient = client;

      client.connect({}, () => {
        state.connected = true;
        setStatus('connected');
        state.subscription = client.subscribe(`/topic/group/${state.backboneGroupId}`, (frame) => {
          try { handleIncoming(JSON.parse(frame.body)); }
          catch (err) { console.warn('Lesson chat: bad frame', err); }
        });
        resolve(true);
      }, (err) => {
        console.warn('Lesson chat: connection failed', err);
        state.connected = false;
        setStatus('disconnected');
        resolve(false);
      });

      socket.onclose = () => {
        if (!state.connected) return;
        state.connected = false;
        setStatus('disconnected');
        inputEl.disabled = true;
        sendBtn.disabled = true;
      };
    });
  }

  function sendMessage(text) {
    if (!state.connected || !state.stompClient || !state.backboneGroupId) return false;
    const payload = {
      context: 'sendMessage',
      groupId: state.backboneGroupId,
      sender: state.currentUserDisplayName || 'anonymous',
      message: `${LESSON_MARKER} ${text}`,
      image: null,
      date: new Date().toISOString(),
    };
    try {
      state.stompClient.send('/app/groups.chat', {}, JSON.stringify(payload));
      return true;
    } catch (err) {
      console.error('Lesson chat: send failed', err);
      return false;
    }
  }

  async function initOnce() {
    if (state.initialized) return;
    state.initialized = true;
    setStatus('loading…');
    await resolveCurrentUser();
    const ok = await resolveBackboneGroup();
    if (!ok) {
      console.warn('Lesson chat: backbone group could not be resolved; see prior warnings for cause');
      setStatus('unavailable');
      appendSystem('Lesson chat is not available right now.');
      return;
    }
    await loadHistory();
    const connected = await connect();
    if (connected) {
      inputEl.disabled = false;
      sendBtn.disabled = false;
      inputEl.focus();
    }
  }

  detailsEl.addEventListener('toggle', () => {
    if (detailsEl.open) initOnce();
  });

  formEl.addEventListener('submit', (e) => {
    e.preventDefault();
    const text = inputEl.value.trim();
    if (!text) return;
    if (sendMessage(text)) {
      inputEl.value = '';
    }
  });

  window.addEventListener('beforeunload', () => {
    try {
      if (state.subscription) state.subscription.unsubscribe();
      if (state.stompClient && state.connected) state.stompClient.disconnect();
    } catch (_) { /* ignore */ }
  });
</script>

          
        </div></div>
    </div>
</div>



<!-- Modals for lesson player -->
<!-- LESSON PLAYER MODALS -->

<!-- Floating hamburger menu (mobile only) -->
<button class="floating-menu-btn" id="floating-menu-btn" style="display: none;">
  <i class="fas fa-bars"></i>
</button>

<!-- Timeline modal overlay -->
<div class="timeline-modal-overlay" id="timeline-modal-overlay"></div>

<!-- Timeline modal (mobile only) -->
<div class="timeline-modal" id="timeline-modal">
  <div class="timeline-modal-header">
    <h3>Course Timeline</h3>
    <button class="timeline-close-btn" id="timeline-close-btn" type="button">
      <i class="fas fa-times"></i>
    </button>
  </div>
  <div class="timeline-modal-content" id="timeline-modal-content">
    <!-- Sprint cards will be populated by JavaScript -->
  </div>
</div>



<script type="module">
  import { pythonURI, fetchOptions, javaURI } from '/assets/js/api/config.js';

  /**
   * ============================================
   * CONFIGURATION & CONSTANTS
   * ============================================
   */
  const COURSE = 'csp';
  const WEEK = '2';
  const PAGE_TITLE = "Onboarding Challenge - Ground 0";
  const CURRENT_LESSON_ID = '/sprint1/challenge/csp/';
  const STORAGE_KEY = `${COURSE}-lesson-completion`;
  const STATUS_AUTO_HIDE_MS = 5000;
  const FILE_SIZE_LIMIT_MB = 10;
  const FRONTMATTER_ASSIGNMENT_ID = null;
  const FRONTMATTER_ASSIGNMENT_NAME = null;
  let CURRENT_ASSIGNMENT_ID = FRONTMATTER_ASSIGNMENT_ID;
  let CURRENT_ASSIGNMENT_NAME = FRONTMATTER_ASSIGNMENT_NAME;
  let ASSIGNMENT_RESOLVE_PROMISE = null;

  /**
   * ============================================
   * SPRINT NAVIGATION: Toggle sprint visibility
   * ============================================
   */
  window.toggleSprint = function (sprintId) {
    const lessons = document.getElementById(`sprint-${sprintId}-lessons`);
    const chevron = document.getElementById(`chevron-${sprintId}`);
    if (!lessons || !chevron) return;

    const isHidden = lessons.style.display === 'none';
    lessons.style.display = isHidden ? 'block' : 'none';
    chevron.classList.toggle('fa-chevron-right', !isHidden);
    chevron.classList.toggle('fa-chevron-down', isHidden);
  };

  /**
   * ============================================
   * COMPLETION DATA LAYER: LocalStorage operations
   * ============================================
   */
  function getCompletionData() {
    try {
      const stored = localStorage.getItem(STORAGE_KEY);
      return stored ? JSON.parse(stored) : {};
    } catch (e) {
      console.error('Error reading completion data:', e);
      return {};
    }
  }

  function saveCompletionData(data) {
    try {
      localStorage.setItem(STORAGE_KEY, JSON.stringify(data));
    } catch (e) {
      console.error('Error saving completion data:', e);
    }
  }

  /**
   * ============================================
   * COMPLETION TOGGLING: Toggle lesson status
   * ============================================
   */
  window.toggleComplete = function () {
    const completionData = getCompletionData();
    const isCompleted = completionData[CURRENT_LESSON_ID] || false;
    completionData[CURRENT_LESSON_ID] = !isCompleted;
    saveCompletionData(completionData);

    syncCompletionUI();
    window.dispatchEvent(new CustomEvent('completionChanged', {
      detail: { itemId: CURRENT_LESSON_ID, completed: !isCompleted }
    }));
  };

  /**
   * ============================================
   * UI UPDATES: Individual responsibility functions
   * ============================================
   */
  function updateCompleteButton(isCompleted) {
    const updateButton = (iconId, textId) => {
      const icon = document.getElementById(iconId);
      const text = document.getElementById(textId);
      if (!icon || !text) return;
      
      icon.classList.toggle('fa-circle', !isCompleted);
      icon.classList.toggle('fa-circle-check', isCompleted);
      text.textContent = isCompleted ? 'Completed' : 'Burndown';
    };
    
    updateButton('complete-icon', 'complete-text');
    updateButton('complete-icon-topbar', 'complete-text-topbar');
  }

  function updateLessonItems(completionData) {
    let completed = 0;
    const lessonItems = document.querySelectorAll('.lesson-item');
    
    lessonItems.forEach(item => {
      const lessonId = item.dataset.lessonId;
      const icon = item.querySelector('.lesson-status-icon');
      const isCompleted = completionData[lessonId] || false;
      
      if (icon) {
        icon.classList.toggle('fa-circle', !isCompleted);
        icon.classList.toggle('fa-circle-check', isCompleted);
        icon.classList.toggle('incomplete', !isCompleted);
        icon.classList.toggle('completed', isCompleted);
      }
      
      if (isCompleted) completed++;
    });
    
    return { total: lessonItems.length, completed };
  }

  function updateProgressBars(completionData) {
    const lessonItems = Array.from(document.querySelectorAll('.lesson-item'));
    const stats = updateLessonItems(completionData);
    
    const sidebarProgress = document.getElementById('sidebar-progress');
    if (sidebarProgress) {
      const sidebarPercent = stats.total > 0 ? (stats.completed / stats.total) * 100 : 0;
      sidebarProgress.style.width = `${sidebarPercent}%`;
    }
    
    const progressCount = document.getElementById('progress-count');
    if (progressCount) {
      progressCount.textContent = `${stats.completed}/${stats.total}`;
    }
    
    const lessonProgress = document.getElementById('lesson-progress');
    if (lessonProgress) {
      const currentIndex = lessonItems.findIndex(item => item.classList.contains('active'));
      const lessonPercent = stats.total > 0 ? ((currentIndex + 1) / stats.total) * 100 : 0;
      lessonProgress.style.width = `${lessonPercent}%`;
    }
  }

  function syncCompletionUI() {
    const completionData = getCompletionData();
    const isCompleted = completionData[CURRENT_LESSON_ID] || false;
    updateCompleteButton(isCompleted);
    updateProgressBars(completionData);
  }

  /**
   * ============================================
   * LESSON NAVIGATION: Move between lessons
   * ============================================
   */
  window.prevLesson = function () {
    const lessonItems = Array.from(document.querySelectorAll('.lesson-item'));
    const currentIndex = lessonItems.findIndex(item => item.classList.contains('active'));
    if (currentIndex > 0) {
      window.location.href = lessonItems[currentIndex - 1].href;
    } else {
      alert('You are on the first lesson!');
    }
  };

  window.nextLesson = function () {
    const lessonItems = Array.from(document.querySelectorAll('.lesson-item'));
    const currentIndex = lessonItems.findIndex(item => item.classList.contains('active'));
    if (currentIndex >= 0 && currentIndex < lessonItems.length - 1) {
      window.location.href = lessonItems[currentIndex + 1].href;
    } else {
      alert('You\'ve reached the last lesson!');
    }
  };

  /**
   * ============================================
   * INITIALIZATION: Player setup
   * ============================================
   */
  function initLessonPlayer() {
    if (document.querySelector('.lesson-player')) {
      document.body.style.margin = '0';
      document.body.style.padding = '0';
    }
  }

  function expandCurrentSprint() {
    const activeLesson = document.querySelector('.lesson-item.active');
    if (!activeLesson) return;
    
    const sprintSection = activeLesson.closest('.sprint-section');
    if (sprintSection) {
      const sprintToggle = sprintSection.querySelector('.sprint-toggle');
      if (sprintToggle) sprintToggle.click();
    }
  }

  /**
   * ============================================
   * CROSS-TAB SYNC: Listen for changes
   * ============================================
   */
  function setupSyncListeners() {
    window.addEventListener('focus', () => {
      if (document.querySelector('.lesson-player')) syncCompletionUI();
    });

    document.addEventListener('visibilitychange', () => {
      if (!document.hidden && document.querySelector('.lesson-player')) syncCompletionUI();
    });

    window.addEventListener('storage', (e) => {
      if (e.key === STORAGE_KEY && document.querySelector('.lesson-player')) {
        syncCompletionUI();
      }
    });

    window.addEventListener('completionChanged', () => {
      if (document.querySelector('.lesson-player')) syncCompletionUI();
    });
  }

  /**
   * ============================================
   * TIMELINE MODAL: Mobile navigation
   * ============================================
   */
  function buildTimelineModal() {
    const completionData = getCompletionData();
    const lessonItems = document.querySelectorAll('.lesson-item');
    const sprintGroups = {};

    lessonItems.forEach(item => {
      const sprintSection = item.closest('.sprint-section');
      if (!sprintSection) return;
      
      const sprintTitle = sprintSection.querySelector('.sprint-title').textContent;
      if (!sprintGroups[sprintTitle]) sprintGroups[sprintTitle] = [];
      
      sprintGroups[sprintTitle].push({
        id: item.dataset.lessonId,
        title: item.querySelector('.lesson-title').textContent,
        url: item.href,
        completed: completionData[item.dataset.lessonId] || false,
        active: item.classList.contains('active')
      });
    });

    return sprintGroups;
  }

  function renderTimelineModal(sprintGroups) {
    const modalContent = document.getElementById('timeline-modal-content');
    if (!modalContent) return;

    modalContent.innerHTML = Object.entries(sprintGroups).map(([sprintTitle, lessons]) => {
      const sprintId = sprintTitle.replace(/\s+/g, '-').toLowerCase();
      const lessonHTML = lessons.map(lesson => `
        <a href="${lesson.url}" class="modal-lesson-item ${lesson.active ? 'active' : ''}">
          <i class="fas ${lesson.completed ? 'fa-circle-check' : 'fa-circle'} modal-lesson-status ${lesson.completed ? 'completed' : 'incomplete'}"></i>
          <span class="modal-lesson-title">${lesson.title}</span>
        </a>
      `).join('');
      
      return `
        <div class="modal-sprint-card">
          <div class="modal-sprint-header">
            <h3 class="modal-sprint-title">${sprintTitle}</h3>
            <button class="modal-sprint-toggle" onclick="toggleModalSprint('${sprintId}')">
              <i class="fas fa-chevron-down"></i> View Lessons
            </button>
          </div>
          <div class="modal-lessons" id="modal-${sprintId}">
            ${lessonHTML}
          </div>
        </div>
      `;
    }).join('');

    // Auto-expand active sprint
    const sprintGroups_arr = Object.values(sprintGroups);
    const activeSprintIndex = sprintGroups_arr.findIndex(lessons => 
      lessons.some(l => l.active)
    );
    
    if (activeSprintIndex !== -1) {
      const activeSprintTitle = Object.keys(sprintGroups)[activeSprintIndex];
      const sprintId = activeSprintTitle.replace(/\s+/g, '-').toLowerCase();
      setTimeout(() => {
        window.toggleModalSprint(sprintId);
      }, 0);
    }
  }

  window.toggleModalSprint = function (sprintId) {
    const lessonsContainer = document.getElementById(`modal-${sprintId}`);
    const toggleBtn = document.querySelector(`[onclick="toggleModalSprint('${sprintId}')"] i`);
    if (!lessonsContainer || !toggleBtn) return;

    lessonsContainer.classList.toggle('expanded');
    toggleBtn.classList.toggle('fa-chevron-down', !lessonsContainer.classList.contains('expanded'));
    toggleBtn.classList.toggle('fa-chevron-up', lessonsContainer.classList.contains('expanded'));
  };

  function setupMobileModal() {
    const floatingBtn = document.getElementById('floating-menu-btn');
    const timelineModal = document.getElementById('timeline-modal');
    const overlay = document.getElementById('timeline-modal-overlay');
    const closeBtn = document.getElementById('timeline-close-btn');
    const modalContent = document.getElementById('timeline-modal-content');

    if (window.innerWidth > 768) {
      if (floatingBtn) floatingBtn.style.display = 'none';
      return;
    }

    if (floatingBtn) floatingBtn.style.display = 'flex';

    const sprintGroups = buildTimelineModal();
    renderTimelineModal(sprintGroups);

    const openModal = () => {
      timelineModal.classList.add('open');
      overlay.classList.add('open');
      document.body.style.overflow = 'hidden';
    };

    const closeModal = () => {
      timelineModal.classList.remove('open');
      overlay.classList.remove('open');
      document.body.style.overflow = '';
    };

    if (floatingBtn) floatingBtn.addEventListener('click', openModal);
    if (closeBtn) closeBtn.addEventListener('click', closeModal);
    if (overlay) overlay.addEventListener('click', closeModal);

    document.addEventListener('keydown', (e) => {
      if (e.key === 'Escape' && timelineModal.classList.contains('open')) closeModal();
    });

    if (modalContent) {
      modalContent.addEventListener('click', (e) => {
        if (e.target.closest('.modal-lesson-item')) closeModal();
      });
    }
  }

  /**
   * ============================================
   * ASSIGNMENT SUBMISSION: API operations
   * ============================================
   */
  async function resolveCurrentUser() {
    const uidRes = await fetch(`${pythonURI}/api/id`, fetchOptions);
    if (!uidRes.ok) throw new Error('Not authenticated. Please log in.');
    const { uid } = await uidRes.json();
    const personRes = await fetch(`${javaURI}/api/person/uid/${encodeURIComponent(uid)}`, fetchOptions);
    if (!personRes.ok) throw new Error('Could not resolve your account in the gradebook.');
    const person = await personRes.json();
    return { uid, personId: person.id };
  }

  async function resolveAssignmentName() {
    if (typeof CURRENT_ASSIGNMENT_NAME === 'string') {
      const normalized = CURRENT_ASSIGNMENT_NAME.trim().toLowerCase();
      if (normalized && normalized !== 'null' && normalized !== 'undefined') return CURRENT_ASSIGNMENT_NAME;
    }
    const assignmentId = await ensureAssignmentId();
    const res = await fetch(`${javaURI}/api/assignments/${assignmentId}`, fetchOptions);
    if (!res.ok) throw new Error('Assignment not found on server.');
    return await res.text();
  }

  async function ensureAssignmentId() {
    const hasAssignment = true;
    if (!hasAssignment) {
      if (!CURRENT_ASSIGNMENT_ID) throw new Error("Assignment ID not configured.");
      return CURRENT_ASSIGNMENT_ID;
    }

    // For assignment pages, always resolve against backend so uploads/submissions
    // target the backend assignment ID, not only the static frontmatter ID.
    if (!ASSIGNMENT_RESOLVE_PROMISE) {
      ASSIGNMENT_RESOLVE_PROMISE = autoCreateAssignmentIfNeeded();
    }
    await ASSIGNMENT_RESOLVE_PROMISE;

    if (!CURRENT_ASSIGNMENT_ID) {
      throw new Error('Could not resolve backend assignment ID. Please log in and refresh.');
    }
    return CURRENT_ASSIGNMENT_ID;
  }

  async function submitLinkToBackend(link, description) {
    const assignmentId = await ensureAssignmentId();
    const { uid, personId } = await resolveCurrentUser();
    const content = { type: 'link', url: link, notes: description };
    const res = await fetch(`${javaURI}/api/submissions/${assignmentId}`, {
      ...fetchOptions,
      method: 'POST',
      body: JSON.stringify({
        assignmentId: assignmentId,
        submitterId: personId,
        isGroupSubmission: false,
        content,
        comment: description,
        isLate: false
      })
    });
    if (!res.ok) {
      const e = await res.json().catch(() => ({}));
      throw new Error(e.error || res.statusText);
    }
    // Dual-write to EshikaP1's grades system
    const assignmentLabel = (typeof CURRENT_ASSIGNMENT_NAME === 'string' && CURRENT_ASSIGNMENT_NAME) ? CURRENT_ASSIGNMENT_NAME : String(assignmentId);
    await fetch(`${javaURI}/api/grades`, {
      ...fetchOptions,
      method: 'POST',
      body: JSON.stringify({ uid, assignment: assignmentLabel, score: null, submission: content, submittedAt: new Date().toISOString() })
    }).catch(() => {}); // non-blocking — don't fail submission if grades write fails
  }

  async function submitFilesToBackend(files, notes) {
    const assignmentId = await ensureAssignmentId();
    const { uid, personId } = await resolveCurrentUser();
    const assignmentName = await resolveAssignmentName();
    for (const file of files) {
      const content = { type: 'file', filename: file.name, notes: notes || '' };
      const fd = new FormData();
      fd.append('assignmentName', assignmentName);
      fd.append('assignmentId', String(assignmentId));
      fd.append('userId', personId);
      fd.append('username', uid);
      fd.append('file', file);
      if (notes) fd.append('notes', notes);
      const r = await fetch(`${javaURI}/api/assignment-submissions/upload`, {
        method: 'POST',
        mode: fetchOptions.mode,
        credentials: fetchOptions.credentials,
        headers: { 'X-Origin': 'client' },
        body: fd
      });
      if (!r.ok) {
        const e = await r.json().catch(() => ({}));
        throw new Error(`Upload failed for ${file.name}: ` + (e.error || r.statusText));
      }
      // Dual-write to EshikaP1's grades system
      const assignmentLabel = (typeof CURRENT_ASSIGNMENT_NAME === 'string' && CURRENT_ASSIGNMENT_NAME) ? CURRENT_ASSIGNMENT_NAME : String(assignmentId);
      await fetch(`${javaURI}/api/grades`, {
        ...fetchOptions,
        method: 'POST',
        body: JSON.stringify({ uid, assignment: assignmentLabel, score: null, submission: content, submittedAt: new Date().toISOString() })
      }).catch(() => {}); // non-blocking
    }
  }

  /**
   * ============================================
   * FILE HANDLING: File operations
   * ============================================
   */
  function fileToBase64(file) {
    return new Promise((resolve, reject) => {
      const reader = new FileReader();
      reader.readAsDataURL(file);
      reader.onload = () => {
        let encoded = reader.result.toString().replace(/^data:(.*,)?/, '');
        if ((encoded.length % 4) > 0) {
          encoded += '='.repeat(4 - (encoded.length % 4));
        }
        resolve(encoded);
      };
      reader.onerror = error => reject(error);
    });
  }

  function formatFileSize(bytes) {
    if (bytes === 0) return '0 Bytes';
    const k = 1024;
    const sizes = ['Bytes', 'KB', 'MB'];
    const i = Math.floor(Math.log(bytes) / Math.log(k));
    return Math.round(bytes / Math.pow(k, i) * 100) / 100 + ' ' + sizes[i];
  }

  /**
   * ============================================
   * STATUS MESSAGES: Display feedback
   * ============================================
   */
  function showStatus(type, message) {
    const status = document.getElementById('submission-status');
    if (!status) return;
    const icon = type === 'success' ? 'check-circle' : type === 'loading' ? 'spinner fa-spin' : 'exclamation-circle';
    status.className = `submission-status ${type}`;
    status.innerHTML = `<i class="fas fa-${icon}"></i> ${message}`;
    status.style.display = 'block';
    if (type !== 'loading') {
      setTimeout(() => {
        status.style.display = 'none';
      }, STATUS_AUTO_HIDE_MS);
    }
  }

  /**
   * ============================================
   * SUBMISSION STORAGE: Local persistence
   * ============================================
   */
  function saveSubmissionLocally(type, data) {
    const submissions = JSON.parse(localStorage.getItem('assignment-submissions') || '{}');
    const lessonId = window.location.pathname;
    if (!submissions[lessonId]) submissions[lessonId] = [];
    submissions[lessonId].push({ type, data, timestamp: new Date().toISOString() });
    localStorage.setItem('assignment-submissions', JSON.stringify(submissions));
  }

  /**
   * ============================================
   * SUBMISSION UI: Tab switching & file display
   * ============================================
   */
  window.switchSubmissionType = function (type) {
    document.querySelectorAll('.submission-tab').forEach(tab => tab.classList.remove('active'));
    event.target.closest('.submission-tab').classList.add('active');
    
    document.querySelectorAll('.submission-form').forEach(form => form.classList.remove('active'));
    document.getElementById(`${type}-form`).classList.add('active');
    
    const status = document.getElementById('submission-status');
    if (status) status.style.display = 'none';
  };

  window.removeFile = function (index) {
    const fileInput = document.getElementById('file-input');
    const fileList = document.getElementById('file-list');
    if (!fileInput || !fileList) return;
    
    const dt = new DataTransfer();
    Array.from(fileInput.files).forEach((file, i) => {
      if (i !== index) dt.items.add(file);
    });
    fileInput.files = dt.files;
    updateFileListDisplay();
  };

  /**
   * ============================================
   * FILE DISPLAY: Render file list
   * ============================================
   */
  function updateFileListDisplay() {
    const fileInput = document.getElementById('file-input');
    const fileList = document.getElementById('file-list');
    const fileSubmitBtn = document.getElementById('file-submit-btn');
    
    if (!fileList) return;
    
    fileList.innerHTML = Array.from(fileInput.files).map((file, index) => `
      <div class="file-item">
        <div class="file-info">
          <i class="fas fa-file"></i>
          <div>
            <div class="file-name">${file.name}</div>
            <div class="file-size">${formatFileSize(file.size)}</div>
          </div>
        </div>
        <button type="button" class="remove-file" onclick="removeFile(${index})">
          <i class="fas fa-times"></i>
        </button>
      </div>
    `).join('');
    
    if (fileSubmitBtn) fileSubmitBtn.disabled = fileInput.files.length === 0;
  }

  /**
   * ============================================
   * FORM SUBMISSION: Handle submissions
   * ============================================
   */
  function setupSubmissionForms() {
    const linkForm = document.getElementById('link-form');
    if (linkForm) {
      linkForm.addEventListener('submit', async function (e) {
        e.preventDefault();
        const link = document.getElementById('submission-link').value.trim();
        const description = document.getElementById('link-description').value.trim();
        const submitBtn = linkForm.querySelector('button[type="submit"]');
        submitBtn.disabled = true;
        showStatus('loading', 'Submitting...');
        try {
          await submitLinkToBackend(link, description);
          saveSubmissionLocally('link', { link, description });
          showStatus('success', 'Link submitted successfully!');
          this.reset();
        } catch (err) {
          showStatus('error', err.message);
        } finally {
          submitBtn.disabled = false;
        }
      });
    }

    const fileForm = document.getElementById('file-form');
    if (fileForm) {
      fileForm.addEventListener('submit', async function (e) {
        e.preventDefault();
        const fileInput = document.getElementById('file-input');
        const notes = document.getElementById('file-notes')?.value?.trim() || '';
        const submitBtn = fileForm.querySelector('button[type="submit"]');
        submitBtn.disabled = true;
        showStatus('loading', 'Uploading...');
        try {
          for (const file of fileInput.files) {
            if (file.size > FILE_SIZE_LIMIT_MB * 1024 * 1024) {
              throw new Error(`${file.name} exceeds ${FILE_SIZE_LIMIT_MB}MB limit.`);
            }
          }
          await submitFilesToBackend(Array.from(fileInput.files), notes);
          saveSubmissionLocally('file', { files: Array.from(fileInput.files).map(f => f.name), notes });
          showStatus('success', `${fileInput.files.length} file(s) submitted successfully!`);
          fileInput.value = '';
          updateFileListDisplay();
          this.reset();
        } catch (err) {
          showStatus('error', err.message);
        } finally {
          submitBtn.disabled = false;
        }
      });
    }
  }

  /**
   * ============================================
   * AUTO-CREATE ASSIGNMENT FROM FRONTMATTER
   * ============================================
   */
  async function autoCreateAssignmentIfNeeded() {
    // For assignment pages, always resolve ID against backend.
    const hasAssignment = true;
    if (!hasAssignment) {
      return; // Not an assignment page
    }
    
    try {
      // Call the Spring API to auto-create assignment
      // This endpoint allows any authenticated user (student, person, teacher, admin)
      const createRes = await fetch(`${javaURI}/api/assignments/auto-create`, {
        ...fetchOptions,
        method: 'POST',
        headers: {
          'Content-Type': 'application/x-www-form-urlencoded'
        },
        body: new URLSearchParams({
          name: PAGE_TITLE || "Onboarding Challenge - Ground 0",
          contentUrl: CURRENT_LESSON_ID,
          description: "Master essential development tools and complete team onboarding through hands-on setup, coding, and collaboration. Begin the process for Project-based Learning (PBL) by launching into a project."
        })
      });
      
      if (createRes.ok) {
        const result = await createRes.json();
        // Update active assignment identity to backend-resolved values.
        CURRENT_ASSIGNMENT_ID = result.id;
        CURRENT_ASSIGNMENT_NAME = result.name;
        window.ASSIGNMENT_ID = result.id;
        window.ASSIGNMENT_NAME = result.name;
        
        // Distinguish between newly created (201) and already existing (200)
        if (createRes.status === 201) {
          console.log('✓ Assignment auto-created with ID:', result.id);
        } else if (createRes.status === 200) {
          console.log('ℹ Assignment already exists with ID:', result.id, '- reusing for this session');
        } else {
          console.log('Assignment resolved with ID:', result.id);
        }
      } else if (createRes.status === 403) {
        console.warn('Not authenticated - assignment creation skipped. Please log in to enable assignments.');
      } else if (createRes.status === 400) {
        const error = await createRes.json();
        console.error('Bad request for auto-create assignment:', error.error);
      } else {
        console.error('Failed to auto-create assignment:', createRes.status, createRes.statusText);
      }
    } catch (error) {
      console.error('Error auto-creating assignment:', error);
    }
  }

  /**
   * ============================================
   * MASTER INITIALIZATION
   * ============================================
   */
  function initializePage() {
    // Lesson player setup
    initLessonPlayer();
    expandCurrentSprint();
    syncCompletionUI();
    setupSyncListeners();
    
    // Mobile enhancements
    setupMobileModal();
    window.addEventListener('resize', () => setupMobileModal());
    
    // Auto-create assignment if needed
    autoCreateAssignmentIfNeeded();
    
    // Assignment submission
    setupSubmissionForms();
    
    // File upload handling
    const fileUploadArea = document.getElementById('file-upload-area');
    const fileInput = document.getElementById('file-input');
    if (fileUploadArea && fileInput) {
      fileUploadArea.addEventListener('click', () => fileInput.click());
      fileUploadArea.addEventListener('dragover', (e) => {
        e.preventDefault();
        fileUploadArea.classList.add('dragover');
      });
      fileUploadArea.addEventListener('dragleave', () => {
        fileUploadArea.classList.remove('dragover');
      });
      fileUploadArea.addEventListener('drop', (e) => {
        e.preventDefault();
        fileUploadArea.classList.remove('dragover');
        fileInput.files = e.dataTransfer.files;
        updateFileListDisplay();
      });
      fileInput.addEventListener('change', () => updateFileListDisplay());
    }
  }

  document.addEventListener('DOMContentLoaded', initializePage);
</script>


                                </div>
            </div>
        </main><footer id="masterFooter" class="site-footer h-card">
    <data class="u-url" href="/"></data>

    <!-- <div class="wrapper">

        <div class="social-links"><ul class="social-media-list"><li>
    <a rel="me" href="" target="_blank" title="">
      <span class="grey fa-brands fa- fa-lg"></span>
    </a>
  </li><li>
    <a rel="me" href="" target="_blank" title="">
      <span class="grey fa-brands fa- fa-lg"></span>
    </a>
  </li><li>
    <a rel="me" href="" target="_blank" title="">
      <span class="grey fa-brands fa- fa-lg"></span>
    </a>
  </li>
  <li>
    <a href="/feed.xml" target="_blank" title="Subscribe to syndication feed">
      <svg class="svg-icon grey" viewbox="0 0 16 16">
        <path d="M12.8 16C12.8 8.978 7.022 3.2 0 3.2V0c8.777 0 16 7.223 16 16h-3.2zM2.194
          11.61c1.21 0 2.195.985 2.195 2.196 0 1.21-.99 2.194-2.2 2.194C.98 16 0 15.017 0
          13.806c0-1.21.983-2.195 2.194-2.195zM10.606
          16h-3.11c0-4.113-3.383-7.497-7.496-7.497v-3.11c5.818 0 10.606 4.79 10.606 10.607z"
        />
      </svg>
    </a>
  </li>
</ul>
</div>

    </div> -->

    <!--<p style="font-style: italic;">OpenAesthetic v3.0rc1.1</p>-->

    <div class="footer-links">
        <div class="footer-previous">
            
        </div>

        <div class="footer-home">
            
        </div>

        <div class="footer-next">
            
        </div>
    </div>

</footer>
<script src="https://cdn.jsdelivr.net/npm/mermaid@10/dist/mermaid.min.js"></script>
<script>
    document.addEventListener('DOMContentLoaded', function() {
        const style = getComputedStyle(document.documentElement);
        const theme = style.getPropertyValue('--theme').trim() || "dark";
        const panel = style.getPropertyValue('--panel-mid').trim() || "#0b1220";
        const text = style.getPropertyValue('--pref-text-color').trim() || "#e6eef8";
        const accent = style.getPropertyValue('--pref-accent-color').trim() || "#3b82f6";
        const fontFamily = style.getPropertyValue('--pref-font-family').trim() || "Arial, sans-serif";
        mermaid.initialize({ 
            startOnLoad: false,
            theme: theme,
            securityLevel: 'loose',
            themeVariables: {
                background: panel,
                primaryColor: panel,
                primaryTextColor: text,
                primaryBorderColor: accent,
                lineColor: text,
                arrowheadColor: text,
                fontFamily: fontFamily
            }
        });
        
        // Convert fenced code blocks to mermaid divs
        document.querySelectorAll('pre code.language-mermaid').forEach((block, index) => {
            const content = block.textContent;
            const wrapper = document.createElement('div');
            wrapper.className = 'mermaid';
            wrapper.id = 'mermaid-' + index;
            wrapper.textContent = content;
            block.parentElement.parentNode.replaceChild(wrapper, block.parentElement);
        });
        
        mermaid.run();
    });
</script></body>
</html>
```
